# DKD Risk Prediction in Type 2 Diabetes
## 05. Figures, Rescue Profiling, SHAP, and Decision Curve Analysis

This notebook reproduces the All of Us main and supplementary figures using the corrected frozen models, corrected test partitions, and CAL-derived fixed-specificity thresholds.

It intentionally excludes external validation. No model retraining, recalibration, feature selection, or TEST-set threshold tuning is performed here.

Outputs:
- Figure 1: ROC, precision-recall, calibration, and net false-negative reduction at the 1-year landmark
- Figure 2: rescued-case feature fingerprints for M2→M3, M3→M4, and M4→M5
- Figure 3: mature-model performance across pre-landmark observation density, social SHAP redistribution, and full-TEST decision curve analysis
- Figure 4: genetics-dominant SHAP redistribution and PRS-level attribution
- Supplementary Figure S1: ROC and PR curves across all corrected landmarks
- Supplementary Figure S2: decision curve analysis across all corrected landmarks
- Supplementary Figure S3: case-control feature gaps across observability tertiles in All of Us
- Supplementary Figure S4: exploratory M4→M5 rescue probability across pre-landmark clinical domain coverage
- SHAP dominance sensitivity at top 10%, 20%, and 30%

**Observability terminology.** Figure 3A uses *observation density*, defined from the total number of pre-landmark measurements across non-kidney general clinical domains. Supplementary Figure S4 uses *domain coverage*, defined as the number of those domains with at least one recorded measurement.

```python
def read_gcs_csv(path):
    with fs.open(path, "r") as f:
        return pd.read_csv(f)

def read_gcs_pickle(path):
    with fs.open(path, "rb") as f:
        return pd.read_pickle(f)

def load_gcs_pickle(path):
    with fs.open(path, "rb") as f:
        return pickle.load(f)

def save_gcs_csv(df, path):
    with fs.open(path, "w") as f:
        df.to_csv(f, index=False)
    print("Saved:", path)

def save_figure_all(fig, stem, *, dpi=600):
    """
    Save publication figures locally as PNG, PDF, and TIFF,
    then upload the same files to the results directory.
    """
    paths = {}
    for ext in ["png", "pdf", "tiff"]:
        local = LOCAL_OUT_DIR / f"{stem}.{ext}"
        if ext == "pdf":
            fig.savefig(local, bbox_inches="tight")
        else:
            fig.savefig(local, dpi=dpi, bbox_inches="tight")
        gcs_path = f"{FIGURE_GCS_DIR}/{local.name}"
        fs.put(str(local), gcs_path)
        paths[ext] = str(local)
        print(f"Saved {ext.upper()}: {local} -> {gcs_path}")
    return paths
```

# 2. Build model-specific data and TEST-aligned prediction objects


```python
def model_features(lm, model_id):
    return (
        feature_manifest.loc[
            feature_manifest["landmark"].eq(lm)
            & feature_manifest["model_id"].eq(model_id)
        ]
        .sort_values("feature_order")["feature"]
        .tolist()
    )

def model_frame(lm, model_id):
    feats = model_features(lm, model_id)
    missing = [c for c in feats if c not in landmark_data[lm].columns]
    if missing:
        raise ValueError(f"{lm}/{model_id}: missing model features: {missing[:10]}")
    return landmark_data[lm][[ID_COL, OUTCOME_COL] + feats].copy()

MODEL_DF = {
    lm: {m: model_frame(lm, m) for m in MODEL_ORDER[lm]}
    for lm in LANDMARKS
}

def threshold_for(lm, model_id):
    row = threshold_df.loc[
        threshold_df["landmark"].eq(lm)
        & threshold_df["model_id"].eq(model_id)
    ]
    if len(row) != 1:
        raise ValueError(f"Threshold row missing/ambiguous for {lm}/{model_id}")
    return float(row["threshold"].iloc[0])

def classification_outcome(y, p, thr):
    y = np.asarray(y).astype(int)
    pred = (np.asarray(p, dtype=float) >= thr).astype(int)
    out = np.empty(len(y), dtype=object)
    out[(y == 1) & (pred == 1)] = "TP"
    out[(y == 1) & (pred == 0)] = "FN"
    out[(y == 0) & (pred == 1)] = "FP"
    out[(y == 0) & (pred == 0)] = "TN"
    return out

def build_landmark_result(lm):
    aligned = {}
    merged = None

    for m in MODEL_ORDER[lm]:
        df = predictions[lm][m].loc[
            predictions[lm][m]["split"].eq("test")
        ].copy()

        df[ID_COL] = pd.to_numeric(df[ID_COL], errors="raise").astype("int64")
        df = df.sort_values(ID_COL).reset_index(drop=True)

        thr = threshold_for(lm, m)
        p = df[PROB_COL].astype(float).to_numpy()
        y = df[PRED_OUTCOME_COL].astype(int).to_numpy()

        one = pd.DataFrame({
            ID_COL: df[ID_COL].to_numpy(),
            "y_true": y,
            f"p_{m}": p,
            f"outcome_{m}": classification_outcome(y, p, thr),
        })

        aligned[m] = one[[ID_COL, "y_true", f"p_{m}"]].copy()

        if merged is None:
            merged = one.copy()
        else:
            merged = merged.merge(
                one[[ID_COL, f"p_{m}", f"outcome_{m}"]],
                on=ID_COL,
                how="inner",
                validate="one_to_one",
            )

    benefit_rows = []
    for a, b in zip(MODEL_ORDER[lm][:-1], MODEL_ORDER[lm][1:]):
        oa = merged[f"outcome_{a}"]
        ob = merged[f"outcome_{b}"]

        fn_tp = int(((oa == "FN") & (ob == "TP")).sum())
        tp_fn = int(((oa == "TP") & (ob == "FN")).sum())
        fp_tn = int(((oa == "FP") & (ob == "TN")).sum())
        tn_fp = int(((oa == "TN") & (ob == "FP")).sum())

        benefit_rows.append({
            "from": a,
            "to": b,
            "FN→TP": fn_tp,
            "TP→FN": tp_fn,
            "FP→TN": fp_tn,
            "TN→FP": tn_fp,
            "Net FN reduction": fn_tp - tp_fn,
            "Net FP reduction": fp_tn - tn_fp,
        })

    return {
        "aligned": aligned,
        "merged_transitions_df": merged,
        "benefit_harm": pd.DataFrame(benefit_rows),
    }

LANDMARK_RESULTS = {
    lm: build_landmark_result(lm)
    for lm in LANDMARKS
}

# ============================================================
# Cumulative M1 -> M5 case rescue at 1y
# ============================================================
merged_1y = LANDMARK_RESULTS["1y"]["merged_transitions_df"]

# All true cases in the 1y test set
cases_1y = merged_1y.loc[merged_1y["y_true"] == 1]
n_cases_1y = len(cases_1y)

# Baseline: cases missed by M1 (M1 false negatives)
m1_missed = cases_1y["outcome_M1"].eq("FN")
n_m1_missed = int(m1_missed.sum())

# Of those, how many are correctly classified by M5?
m1_fn_m5_tp = cases_1y.loc[m1_missed, "outcome_M5"].eq("TP")
n_recovered = int(m1_fn_m5_tp.sum())

# Cases still missed by M5 despite M5's additional layers
m5_still_missed = cases_1y["outcome_M5"].eq("FN")
n_m5_still_missed = int(m5_still_missed.sum())

# Cumulative recovery rate
pct_recovered = 100 * n_recovered / n_m1_missed if n_m1_missed else float("nan")

print("=" * 60)
print("Cumulative M1 -> M5 case rescue at the 1y landmark (test set)")
print("=" * 60)
print(f"Total cases (test set):              {n_cases_1y:,}")
print(f"Cases missed by M1 (M1 FN):          {n_m1_missed:,}")
print(f"  ...correctly detected by M5 (TP):  {n_recovered:,}  ({pct_recovered:.1f}%)")
print(f"Cases still missed by M5:            {n_m5_still_missed:,}")
print()

# Sanity check: also compute cumulative "net" the way the paper's original 30.4% was framed
# Original paper: net FN reduction over M1 = (M1 FN) - (M5 FN)
net_reduction = n_m1_missed - n_m5_still_missed
pct_net_reduction = 100 * net_reduction / n_m1_missed if n_m1_missed else float("nan")
print(f"Net M1 FN reduction (M1 FN - M5 FN): {net_reduction:,}  ({pct_net_reduction:.1f}%)")
res_1y = LANDMARK_RESULTS["1y"]

for lm in LANDMARKS:
    print(f"\n{lm}")
    print(LANDMARK_RESULTS[lm]["benefit_harm"].to_string(index=False))
```

# 3. Figure 1 — 1-year discrimination, calibration, and net reduction in missed cases


```python
# ============================================================
# Figure 1 (2x2): ROC, PR, Calibration, Waterfall
# - Works for chains like M1->M2->M3->M4->M5 OR ...->M4a->M5
# - Waterfall can use:
#     * net (FN→TP - TP→FN)  [recommended]
#     * rescued (FN→TP only) [gross rescued]
# - Legends placed INSIDE panels (2x2-friendly)
# - PR y-min defaults to prevalence (e.g., 0.41) for cleaner scale
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
from sklearn.metrics import roc_curve, precision_recall_curve, roc_auc_score, average_precision_score
from sklearn.calibration import calibration_curve


# ----------------------------
# 1) Global style (removes default matplotlib feel)
# ----------------------------
def set_pub_rc():
    plt.rcParams.update({
        "font.size": 10.5,
        "axes.titlesize": 11,
        "axes.labelsize": 10.5,
        "legend.fontsize": 8.6,
        "xtick.labelsize": 9.0,
        "ytick.labelsize": 9.0,
        "axes.linewidth": 0.9,
        "lines.linewidth": 1.6,
        "savefig.bbox": "tight",
    })


def _remove_axes_box(ax):
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_linewidth(0.9)
    ax.spines["bottom"].set_linewidth(0.9)


def _soft_grid(ax, which="both"):
    if which == "y":
        ax.grid(axis="y", alpha=0.12, linewidth=0.8)
    else:
        ax.grid(True, alpha=0.12, linewidth=0.8)


# ----------------------------
# 2) High-contrast, journal-friendly palette
#    M4a is treated as the same layer as M4 visually
# ----------------------------
PALETTE = {
    "M1": "#6E6E6E",  # gray
    "M2": "#3B82F6",  # blue (distinct)
    "M3": "#10B981",  # green (distinct from blue)
    "M4": "#1E293B",  # navy
    "M4a": "#1E293B", # navy
    "M5": "#E76F51",  # warm orange (final emphasis)
}

# Line hierarchy:
# - M1 dashed baseline
# - M5 thicker emphasis
CURVE_STYLE = {
    "M1": dict(ls="--", lw=1.8, alpha=0.95),
    "M2": dict(ls="-",  lw=1.7, alpha=0.95),
    "M3": dict(ls="-",  lw=1.7, alpha=0.95),
    "M4": dict(ls="-",  lw=1.7, alpha=0.95),
    "M4a": dict(ls="-", lw=1.7, alpha=0.95),
    "M5": dict(ls="-",  lw=2.4, alpha=1.00),
}


# ============================================================
# 3) Infer the evaluated model chain from benefit_harm
# ============================================================
def infer_model_order_from_benefit_harm(res):
    """
    Reconstruct chain order from benefit_harm edges:
      from -> to
    Handles M4a automatically.
    """
    bh = res["benefit_harm"].copy()
    edges = list(zip(bh["from"].astype(str), bh["to"].astype(str)))
    froms = [a for a, _ in edges]
    tos   = [b for _, b in edges]

    # start node = appears in 'from' but not in 'to'
    start = next((x for x in froms if x not in set(tos)), froms[0])

    # follow the chain
    nxt = {a: b for a, b in edges}
    order = [start]
    while order[-1] in nxt:
        order.append(nxt[order[-1]])

    return tuple(order)


# ============================================================
# 4) Waterfall inputs from benefit_harm + merged_transitions_df
# ============================================================
def rescued_from_benefit_harm(
    res,
    model_order,
    *,
    waterfall_mode="net",   # "net" or "rescued"
):
    """
    Build the waterfall spec from the exact evaluation outputs.

    Start FN:
      - computed from merged_transitions_df outcome_<first_model> == "FN"
        (guaranteed consistent with benefit_harm transitions)

    Steps:
      - if waterfall_mode == "net":
          Net FN reduction = (FN→TP) - (TP→FN)
      - if waterfall_mode == "rescued":
          FN→TP only (gross rescued)

    Remaining:
      start_fn - sum(steps)   (net accounting)
    """
    bh = res["benefit_harm"].copy()
    mt = res["merged_transitions_df"].copy()

    first = model_order[0]
    if f"outcome_{first}" not in mt.columns:
        raise ValueError(f"merged_transitions_df missing outcome_{first}")

    # Start FN computed from outcomes in merged_transitions_df (most consistent)
    start_fn = int((mt[f"outcome_{first}"].astype(str) == "FN").sum())

    step_vals = []
    for a, b in zip(model_order[:-1], model_order[1:]):
        row = bh[(bh["from"].astype(str) == a) & (bh["to"].astype(str) == b)]
        if row.shape[0] != 1:
            raise ValueError(f"Missing or ambiguous benefit_harm row for {a}->{b}")

        if waterfall_mode == "net":
            v = int(row["Net FN reduction"].iloc[0])
        elif waterfall_mode == "rescued":
            v = int(row["FN→TP"].iloc[0])
        else:
            raise ValueError("waterfall_mode must be 'net' or 'rescued'")

        step_vals.append(v)

    still_fn = start_fn - int(np.sum(step_vals)) if waterfall_mode == "net" else None

    return {
        "start_fn": start_fn,
        "steps": step_vals,
        "still_fn": still_fn,        # only meaningful for "net"
        "model_order": model_order,
        "waterfall_mode": waterfall_mode,
    }


# ============================================================
# 5a) ROC panel
# ============================================================
def plot_roc_on_ax(ax, aligned, *, order, xpad=0.02):
    # faint diagonal reference (very light)
    ax.plot([0, 1], [0, 1], linestyle=":", linewidth=0.7, color="0.92", zorder=1)

    for i, k in enumerate(order):
        df = aligned[k]
        y = df["y_true"].to_numpy().astype(int)
        p = df[f"p_{k}"].to_numpy().astype(float)

        fpr, tpr, _ = roc_curve(y, p)
        auc = roc_auc_score(y, p)
        st = CURVE_STYLE.get(k, dict(ls="-", lw=1.7, alpha=0.95))

        ax.plot(
            fpr, tpr,
            color=PALETTE.get(k, None),
            linestyle=st["ls"],
            linewidth=st["lw"],
            alpha=st["alpha"],
            solid_capstyle="round",
            dash_capstyle="round",
            zorder=10 + i,
            label=f"{k} (AUC {auc:.3f})",
        )

    ax.set_xlabel("False Positive Rate")
    ax.set_ylabel("True Positive Rate")
    ax.set_xlim(-xpad, 1.0)
    ax.set_ylim(0, 1.0)
    _soft_grid(ax)
    _remove_axes_box(ax)

    # legend INSIDE (2x2-friendly)
    ax.legend(loc="lower right", frameon=False, handlelength=2.6, handletextpad=0.6)


# ============================================================
# 5b) PR panel
# ============================================================
def plot_pr_on_ax(ax, aligned, *, order, ymin=None):
    y0 = aligned[order[0]]["y_true"].to_numpy().astype(int)
    prev = float(np.mean(y0))

    # default ymin = prevalence (or slightly below)
    if ymin is None:
        ymin = max(0.0, prev - 0.01)

    # very light prevalence line
    ax.axhline(prev, linestyle=":", linewidth=0.8, color="0.88", zorder=1)
    ax.text(0.98, prev, f"prevalence = {prev:.2f}", ha="right", va="bottom", fontsize=8, color="0.45")

    for i, k in enumerate(order):
        df = aligned[k]
        y = df["y_true"].to_numpy().astype(int)
        p = df[f"p_{k}"].to_numpy().astype(float)

        prec, rec, _ = precision_recall_curve(y, p)
        ap = average_precision_score(y, p)
        st = CURVE_STYLE.get(k, dict(ls="-", lw=1.7, alpha=0.95))

        ax.plot(
            rec, prec,
            color=PALETTE.get(k, None),
            linestyle=st["ls"],
            linewidth=st["lw"],
            alpha=st["alpha"],
            solid_capstyle="round",
            dash_capstyle="round",
            zorder=10 + i,
            label=f"{k} (AUPRC {ap:.3f})",
        )

    ax.set_xlabel("Recall")
    ax.set_ylabel("Precision")
    ax.set_xlim(0, 1)
    ax.set_ylim(ymin, 1.0)
    _soft_grid(ax)
    _remove_axes_box(ax)

    # legend INSIDE
    ax.legend(loc="upper right", frameon=False, handlelength=2.6, handletextpad=0.6)


# ============================================================
# 5c) Calibration panel (quiet)
# ============================================================
def plot_calibration_on_ax(ax, aligned, *, order, n_bins=10, strategy="uniform"):
    ax.plot([0, 1], [0, 1], linestyle=":", linewidth=0.7, color="0.92", zorder=1)

    for i, k in enumerate(order):
        df = aligned[k]
        y = df["y_true"].to_numpy().astype(int)
        p = df[f"p_{k}"].to_numpy().astype(float)

        obs, pred = calibration_curve(y, p, n_bins=n_bins, strategy=strategy)

        # keep calibration visually quiet: thinner lines, no markers
        lw = 0.95 if k != "M5" else 1.25
        alpha = 0.85 if k != "M5" else 0.95
        ls = "--" if k == "M1" else "-"

        ax.plot(
            pred, obs,
            color=PALETTE.get(k, None),
            linewidth=lw,
            alpha=alpha,
            linestyle=ls,
            zorder=5 + i,
            label=k,
        )

    ax.set_xlabel("Predicted probability")
    ax.set_ylabel("Observed frequency")
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    _soft_grid(ax)
    _remove_axes_box(ax)

    # legend INSIDE
    ax.legend(loc="upper left", frameon=False, handlelength=2.6, handletextpad=0.6)


# ============================================================
# 5d) Waterfall panel
# ============================================================
def plot_waterfall_on_ax(ax, wf, *, show_percent=True):
    """
    Waterfall is shown as reductions in missed cases.

    If wf["waterfall_mode"] == "net":
      - Start: FN under first model
      - Gains: -Net FN reduction per step
      - Remaining: FN after all steps
      - % shown relative to Start FN (explicit denominator)

    If wf["waterfall_mode"] == "rescued":
      - Start: FN under first model
      - Gains: -FN→TP per step (gross rescued)
      - Remaining is not shown (because accounting is not net)
        (You can add a final bar if you want, but it won't match net identity.)
    """
    model_order = wf["model_order"]
    steps = np.array(wf["steps"], dtype=float)
    start_fn = float(wf["start_fn"])
    mode = wf["waterfall_mode"]

    labels = [f"Start ({model_order[0]})"] + [f"{b} gain" for b in model_order[1:]]
    colors = [PALETTE.get(model_order[0], "#6E6E6E")] + [PALETTE.get(m, "#1E293B") for m in model_order[1:]]

    # If net: append "Remaining"
    if mode == "net":
        still_fn = float(wf["still_fn"])
        labels = labels + ["Remaining"]
        colors = colors + [PALETTE["M5"]]
        heights = np.array([start_fn, *(-steps), still_fn], dtype=float)
    else:
        heights = np.array([start_fn, *(-steps)], dtype=float)

    x = np.arange(len(labels))

    # bottoms: waterfall style where each reduction starts at previous level
    bottoms = np.zeros_like(heights)
    levels = [start_fn]
    for d in steps:
        levels.append(levels[-1] - d)
    if mode == "net":
        levels.append(still_fn)
    levels = np.array(levels, dtype=float)

    for i in range(1, 1 + len(steps)):
        bottoms[i] = levels[i - 1]
    if mode == "net":
        bottoms[-1] = 0.0  # remaining bar from 0

    bar_width = 0.80
    ax.bar(x, heights, bottom=bottoms, width=bar_width, color=colors, edgecolor="none")

    # connectors (subtle)
    for i in range(len(steps)):
        yline = levels[i]
        ax.plot(
            [x[i] + bar_width / 2, x[i + 1] - bar_width / 2],
            [yline, yline],
            color="0.30",
            linewidth=0.9,
            solid_capstyle="round",
            zorder=5
        )

    ax.set_ylabel("Missed DKD cases (false negatives)")
    ax.set_xticks(x)
    ax.set_xticklabels(labels, fontsize=8.8)
    ax.yaxis.set_major_formatter(FuncFormatter(lambda v, pos: f"{int(v):,}"))
    ax.tick_params(axis="y", labelsize=9.0)
    ax.grid(axis="y", alpha=0.12, linewidth=0.8)
    _remove_axes_box(ax)

    ax.set_ylim(0, start_fn * 1.12)

    # annotations
    ypad_top = start_fn * 0.02
    ypad_under = start_fn * 0.02

    ax.text(x[0], start_fn + ypad_top, f"{int(start_fn):,}", ha="center", va="bottom",
            fontsize=9.0, color="0.10", fontweight="semibold")

    if mode == "net":
        ax.text(x[-1], still_fn + ypad_top, f"{int(still_fn):,}", ha="center", va="bottom",
                fontsize=9.0, color="0.10", fontweight="semibold")

    for i, d in enumerate(steps, start=1):
        pct = (d / start_fn * 100.0) if start_fn > 0 else np.nan
        txt = f"−{int(d):,}"
        if show_percent and np.isfinite(pct):
            # explicit denominator text to avoid reviewer confusion
            txt += f"\n({pct:.1f}%)\nof {model_order[0]} FN"

        y_after_drop = levels[i]
        ax.text(
            x[i], y_after_drop - ypad_under, txt,
            ha="center", va="top",
            fontsize=8.3, color="0.10", fontweight="semibold"
        )


# ============================================================
# 6) MASTER: Build Figure 1 from res
# ============================================================
def make_figure1_from_res(
    res,
    *,
    waterfall_mode="net",   # "net" (recommended) or "rescued"
    pr_ymin=None,           # default None -> prevalence-based
    roc_xpad=0.02,
    calib_bins=10,
    figsize=(11.0, 8.0),
    dpi=600,
    outpath_png=None,
):
    set_pub_rc()

    aligned = res["aligned"]

    # Infer order from benefit_harm so baseline uses M4a automatically
    model_order = infer_model_order_from_benefit_harm(res)

    # Keep only models present in aligned (safety)
    model_order = tuple([k for k in model_order if k in aligned])

    # Build waterfall spec directly from evaluation tables
    wf = rescued_from_benefit_harm(res, model_order, waterfall_mode=waterfall_mode)

    fig, axes = plt.subplots(2, 2, figsize=figsize, dpi=dpi)
    ax_a, ax_b = axes[0, 0], axes[0, 1]
    ax_c, ax_d = axes[1, 0], axes[1, 1]

    # a) ROC
    plot_roc_on_ax(ax_a, aligned, order=model_order, xpad=roc_xpad)

    # b) PR (ymin defaults to prevalence)
    plot_pr_on_ax(ax_b, aligned, order=model_order, ymin=pr_ymin)

    # c) Calibration
    plot_calibration_on_ax(ax_c, aligned, order=model_order, n_bins=calib_bins, strategy="uniform")

    # d) Waterfall
    plot_waterfall_on_ax(ax_d, wf, show_percent=True)

    # panel letters
    for label, ax in zip(["a", "b", "c", "d"], [ax_a, ax_b, ax_c, ax_d]):
        ax.text(-0.12, 1.06, label, transform=ax.transAxes,
                fontsize=14, fontweight="bold", va="top", ha="left")

    fig.tight_layout()

    if outpath_png is not None:
        fig.savefig(outpath_png, dpi=dpi, bbox_inches="tight")

    return fig, axes, wf
```

```python
import numpy as np
import pandas as pd

# ============================================================
# Helpers: inference + SMD
# ============================================================

def infer_binary_cols(df: pd.DataFrame, cols=None, max_unique: int = 3) -> set[str]:
    if cols is None:
        cols = list(df.columns)
    out = set()
    for c in cols:
        if c not in df.columns:
            continue
        s = df[c].dropna()
        if s.empty:
            continue
        u = set(pd.unique(s))
        if len(u) <= max_unique and u.issubset({0, 1, 0.0, 1.0, True, False}):
            out.add(c)
    return out

def _safe_float(x):
    try:
        return float(x)
    except Exception:
        return np.nan

def smd_continuous(xA: np.ndarray, xB: np.ndarray, *, ddof: int = 1) -> float:
    xA = np.asarray(xA, dtype=float)
    xB = np.asarray(xB, dtype=float)
    xA = xA[~np.isnan(xA)]
    xB = xB[~np.isnan(xB)]
    if xA.size < 2 or xB.size < 2:
        return np.nan
    meanA = np.mean(xA); meanB = np.mean(xB)
    varA = np.var(xA, ddof=ddof); varB = np.var(xB, ddof=ddof)
    pooled = np.sqrt((varA + varB) / 2.0)
    if not np.isfinite(pooled) or pooled == 0:
        return np.nan
    return (meanA - meanB) / pooled

def smd_binary(xA: np.ndarray, xB: np.ndarray) -> float:
    a = np.asarray(xA, dtype=float)
    b = np.asarray(xB, dtype=float)
    a = a[~np.isnan(a)]
    b = b[~np.isnan(b)]
    if a.size == 0 or b.size == 0:
        return np.nan
    a = (a > 0).astype(float)
    b = (b > 0).astype(float)
    pA = np.mean(a); pB = np.mean(b)
    pooled = np.sqrt((pA * (1 - pA) + pB * (1 - pB)) / 2.0)
    if not np.isfinite(pooled) or pooled == 0:
        return np.nan
    return (pA - pB) / pooled

def _feature_type(df: pd.DataFrame, col: str, binary_features: set[str]) -> str:
    if col in binary_features:
        return "binary"
    if pd.api.types.is_bool_dtype(df[col]):
        return "binary"
    s = df[col].dropna()
    if not s.empty:
        u = set(pd.unique(s))
        if len(u) <= 3 and u.issubset({0, 1, 0.0, 1.0, True, False}):
            return "binary"
    return "continuous"

def profile_compare_features(
    dfA: pd.DataFrame,
    dfB: pd.DataFrame,
    features_bcols: list[str],
    *,
    binary_features: set[str] | None = None,
    min_n_cont: int = 25,
    min_n_bin: int = 25,
    min_prev: float = 0.01,
) -> pd.DataFrame:
    binary_features = binary_features or set()
    rows = []
    combined = pd.concat([dfA, dfB], axis=0)

    for f in features_bcols:
        if f not in dfA.columns or f not in dfB.columns:
            continue

        ftype = _feature_type(combined[[f]], f, binary_features)

        xA = dfA[f].to_numpy()
        xB = dfB[f].to_numpy()
        nA = int(np.sum(~pd.isna(xA)))
        nB = int(np.sum(~pd.isna(xB)))

        if ftype == "continuous":
            if nA < min_n_cont or nB < min_n_cont:
                continue
            smd = smd_continuous(xA, xB)
            if not np.isfinite(smd):
                continue
            rows.append({
                "feature_b": f,
                "type": "continuous",
                "smd_A_minus_B": float(smd),
                "|smd|": float(abs(smd)),
                "nA_nonmiss": nA,
                "nB_nonmiss": nB,
                "meanA": _safe_float(np.nanmean(xA.astype(float))),
                "meanB": _safe_float(np.nanmean(xB.astype(float))),
                "prevA": np.nan,
                "prevB": np.nan,
            })
        else:
            if nA < min_n_bin or nB < min_n_bin:
                continue
            a01 = (pd.Series(xA).astype(float) > 0).astype(float)
            b01 = (pd.Series(xB).astype(float) > 0).astype(float)
            pA = _safe_float(np.nanmean(a01))
            pB = _safe_float(np.nanmean(b01))
            if (np.isfinite(pA) and pA < min_prev) and (np.isfinite(pB) and pB < min_prev):
                continue
            smd = smd_binary(a01.to_numpy(), b01.to_numpy())
            if not np.isfinite(smd):
                continue
            rows.append({
                "feature_b": f,
                "type": "binary",
                "smd_A_minus_B": float(smd),
                "|smd|": float(abs(smd)),
                "nA_nonmiss": nA,
                "nB_nonmiss": nB,
                "meanA": np.nan,
                "meanB": np.nan,
                "prevA": pA,
                "prevB": pB,
            })

    out = pd.DataFrame(rows)
    if out.empty:
        return out
    return out.sort_values("|smd|", ascending=False).reset_index(drop=True)


# ============================================================
# Cohorts: rescued vs still_FN (uses merged_preds outcomes)
# ============================================================

def build_transition_masks(merged_preds: pd.DataFrame, *, a: str, b: str) -> dict[str, pd.Series]:
    oa = merged_preds[f"outcome_{a}"].astype(str)
    ob = merged_preds[f"outcome_{b}"].astype(str)
    rescued = (oa.eq("FN") & ob.eq("TP"))
    still_FN = (oa.eq("FN") & ob.eq("FN"))
    return {"rescued_FN_to_TP": rescued, "still_FN": still_FN}


# ============================================================
# Canonicalization to handle snapshot suffix mismatch
# ============================================================

SNAPSHOT_BASES = {
    "SBP","DBP","HR",
    "HbA1c","FBG","RBG",
    "serum_creatinine","Screatinine","eGFR","UACR","UrAlb",
    "LDL","HDL","TG","TotalChol",
    "Na","K","Cl","HCO3","Mg","BUN",
    "ALT","AST","ALP","Albumin","TotalProtein",
    "TotalBilirubin","DirectBilirubin",
    "Hgb","ESR",
    "Weight","Height","BMI",
}

def canonicalize_feature(model_key: str, col: str) -> str:
    if col.endswith("_last"):
        base = col[:-5]
        if base in SNAPSHOT_BASES:
            return base
    return col

def build_canonical_maps(model_key: str, cols: list[str], *, id_col: str, outcome_col: str) -> dict[str, str]:
    out = {}
    for c in cols:
        if c in {id_col, outcome_col}:
            continue
        canon = canonicalize_feature(model_key, c)
        if canon not in out:
            out[canon] = c
    return out


# ============================================================
# Main runner: y1-friendly
# ============================================================

def run_pairwise_rescue_profiling_y1(
    *,
    merged_preds: pd.DataFrame,                     # MUST include outcome_<model> and person_id
    model_df: Dict[str, pd.DataFrame],              # {"M1": df, "M2": df, ...} raw feature DFs
    transitions: list[tuple[str, str]],             # [("M1","M2"), ...]
    id_col: str = "person_id",
    outcome_col: str = "dkd_binary",
    control_group: str = "still_FN",
    top_k_delta: int = 15,
    top_k_combined: int = 50,
    min_n_cont: int = 25,
    min_n_bin: int = 25,
    min_prev: float = 0.01,
    binary_overrides: set[str] | None = None,
    delta_canon_override: dict[tuple[str, str], set[str]] | None = None,
) -> dict:
    """
    For each a→b:
      - cohort: rescued FN→TP vs still_FN (both defined from merged_preds outcomes)
      - feature source: df_b (raw b-space features)
      - shared/new determined via canonical names (helps with _last suffix mismatch)
      - outputs delta-only and combined tables ranked by |SMD|
    """
    delta_canon_override = delta_canon_override or {}

    # MODEL_COLS from the provided DFs
    model_cols = {k: v.columns.tolist() for k, v in model_df.items()}

    # infer binary overrides once from widest DF if not provided
    if binary_overrides is None:
        widest_key = max(model_df.keys(), key=lambda k: model_df[k].shape[1])
        df_for_bin = model_df[widest_key]
        inferred = infer_binary_cols(df_for_bin, cols=[c for c in df_for_bin.columns if c != id_col])
        suffix_forced = {
            c for c in df_for_bin.columns
            if c.endswith(("_ever","_recent_30d","_missing","_skip","_pna","_frac_abnormal","_ever_abnormal"))
        }
        binary_overrides = inferred | suffix_forced

    keep_ids = set(pd.to_numeric(merged_preds[id_col], errors="coerce").dropna().astype("int64").unique().tolist())

    results = {
        "cohort_sizes": [],
        "per_transition": {},
        "meta": {
            "top_k_delta": top_k_delta,
            "top_k_combined": top_k_combined,
            "control_group": control_group,
            "binary_overrides_n": int(len(binary_overrides)),
        }
    }

    for (a, b) in transitions:
        key = f"{a}_to_{b}"
        results["per_transition"][key] = {}

        masks = build_transition_masks(merged_preds, a=a, b=b)
        size_row = {"from": a, "to": b}
        for cname, m in masks.items():
            size_row[cname] = int(m.sum())
        results["cohort_sizes"].append(size_row)

        idsA = pd.to_numeric(merged_preds.loc[masks["rescued_FN_to_TP"], id_col], errors="coerce").dropna().astype("int64").to_numpy()
        idsB = pd.to_numeric(merged_preds.loc[masks[control_group], id_col], errors="coerce").dropna().astype("int64").to_numpy()

        # feature source: df_b raw
        df_b = model_df[b].copy()
        df_b[id_col] = pd.to_numeric(df_b[id_col], errors="coerce").astype("Int64")
        df_b = df_b.dropna(subset=[id_col])
        df_b[id_col] = df_b[id_col].astype("int64")
        df_b = df_b[df_b[id_col].isin(keep_ids)].drop_duplicates(subset=[id_col], keep="first").set_index(id_col)

        dfA = df_b.loc[df_b.index.intersection(idsA)].copy()
        dfB = df_b.loc[df_b.index.intersection(idsB)].copy()

        if dfA.shape[0] == 0 or dfB.shape[0] == 0:
            results["per_transition"][key]["delta_table"] = pd.DataFrame()
            results["per_transition"][key]["combined_table"] = pd.DataFrame()
            results["per_transition"][key]["new_features_b"] = []
            results["per_transition"][key]["shared_features_b"] = []
            continue

        canonA = build_canonical_maps(a, model_cols[a], id_col=id_col, outcome_col=outcome_col)
        canonB = build_canonical_maps(b, model_cols[b], id_col=id_col, outcome_col=outcome_col)

        setA = set(canonA.keys())
        setB = set(canonB.keys())

        shared_canon = sorted(list(setA & setB))
        new_canon = sorted(list(setB - setA))

        override = delta_canon_override.get((a, b))
        if override is not None:
            new_canon = sorted([c for c in override if c in setB])
            shared_canon = sorted(list((setA & setB) - set(new_canon)))

        shared_b = [canonB[c] for c in shared_canon if canonB[c] in dfA.columns and canonB[c] in dfB.columns]
        new_b = [canonB[c] for c in new_canon if canonB[c] in dfA.columns and canonB[c] in dfB.columns]

        delta_tbl = pd.DataFrame()
        if len(new_b) > 0:
            delta_tbl = profile_compare_features(
                dfA=dfA, dfB=dfB, features_bcols=new_b,
                binary_features=binary_overrides,
                min_n_cont=min_n_cont, min_n_bin=min_n_bin, min_prev=min_prev,
            )
            if not delta_tbl.empty:
                delta_tbl = delta_tbl.head(top_k_delta).copy()

        combined_cols = list(dict.fromkeys(new_b + shared_b))
        combined_tbl = pd.DataFrame()
        if len(combined_cols) > 0:
            combined_tbl = profile_compare_features(
                dfA=dfA, dfB=dfB, features_bcols=combined_cols,
                binary_features=binary_overrides,
                min_n_cont=min_n_cont, min_n_bin=min_n_bin, min_prev=min_prev,
            )
            if not combined_tbl.empty:
                new_set = set(new_b)
                combined_tbl["kind"] = combined_tbl["feature_b"].apply(lambda x: "new" if x in new_set else "shared")
                combined_tbl = combined_tbl.head(top_k_combined).copy()

        results["per_transition"][key]["new_features_b"] = new_b
        results["per_transition"][key]["shared_features_b"] = shared_b
        results["per_transition"][key]["delta_table"] = delta_tbl
        results["per_transition"][key]["combined_table"] = combined_tbl

    results["cohort_sizes"] = pd.DataFrame(results["cohort_sizes"])
    return results


def pretty_print_pairwise_rescue(results: dict, *, transitions: list[tuple[str,str]]):
    cs = results["cohort_sizes"]
    print("\n" + "="*90)
    print("COHORT SIZES PER TRANSITION")
    print("="*90)
    print(cs.to_string(index=False))

    base_cols = ["feature_b","kind","type","smd_A_minus_B","|smd|","nA_nonmiss","nB_nonmiss","meanA","meanB","prevA","prevB"]
    delta_cols = ["feature_b","type","smd_A_minus_B","|smd|","nA_nonmiss","nB_nonmiss","meanA","meanB","prevA","prevB"]

    for (a,b) in transitions:
        key = f"{a}_to_{b}"
        blob = results["per_transition"][key]

        print("\n" + "="*90)
        print(f"TRANSITION: {a} -> {b}  (rescued FN->TP vs still_FN)")
        print("="*90)

        print(f"\nNew features in {b} (b-space columns): n={len(blob['new_features_b'])}")
        if len(blob["new_features_b"]) == 0:
            print("(none)")
        elif len(blob["new_features_b"]) <= 35:
            print(", ".join(blob["new_features_b"]))
        else:
            print(", ".join(blob["new_features_b"][:35]) + " ...")

        print("\n" + "-"*90)
        print("NEW-FEATURE DIFFERENCES (delta only, ranked by |SMD|)")
        print("-"*90)
        dt = blob["delta_table"]
        if dt is None or dt.empty:
            print("(none met thresholds)")
        else:
            print(dt[delta_cols].to_string(index=False))

        print("\n" + "-"*90)
        print("COMBINED DIFFERENCES (new + shared, ranked by |SMD|)")
        print("-"*90)
        ct = blob["combined_table"]
        if ct is None or ct.empty:
            print("(none met thresholds)")
        else:
            cols = [c for c in base_cols if c in ct.columns]
            print(ct[cols].to_string(index=False))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Social feature rules (same idea you used in model builders)
# -----------------------------
SOCIAL_PREFIXES = (
    "education_", "income_", "housing_", "employment_status_",
    "insurance_status_", "insurance_type_",
    "race_", "home_", "living_", "marital_",
    "lonely_", "loneliness_", "mistrust_", "harass_", "harassment_",
    "stress_", "disorder_", "neighborhood_",
    "Alcohol_", "Cigar_", "Electronic_", "Hookah_",
    "Past_3_Month_", "Smokeless_", "Smoking_",
    "rec_",
)
SOCIAL_EXACT = {
    "education_level", "income_level", "housing_unstable", "living_missing",
    "loneliness_mean", "mistrust_mean", "stress_mean",
    "harassment_score_mean", "neighborhood_disorder_mean",
}

def is_social_feature(c: str) -> bool:
    return (c in SOCIAL_EXACT) or c.startswith(SOCIAL_PREFIXES)

def _prettify_social_name(f: str) -> str:
    """
    Make long survey names more readable on the y-axis.
    Keep it conservative: only light cleanup.
    """
    f = f.replace("employment_status_", "employment:")
    f = f.replace("insurance_status_", "insurance status:")
    f = f.replace("insurance_type_", "insurance type:")
    f = f.replace("race_", "race:")
    f = f.replace("marital_", "marital:")
    f = f.replace("living_", "living:")
    f = f.replace("home_", "home:")
    f = f.replace("Alcohol_", "Alcohol:")
    f = f.replace("Smoking_", "Smoking:")
    f = f.replace("Smokeless_Tobacco_", "Smokeless:")
    f = f.replace("Past_3_Month_Use_Frequency_", "Past 3mo:")
    f = f.replace("_Participant", "")
    f = f.replace("_Occurrence", "")
    f = f.replace("_Frequency", " frequency")
    f = f.replace("_Past_Year", " past year")
    f = f.replace("_", " ")
    return f
```

```python
import re
import numpy as np
import matplotlib.pyplot as plt
import textwrap

# ============================================================
# Figure 2: Fingerprint plots (1x3)
#  a) Medication signals driving rescue (M2 → M3)
#  b) Monitoring signals driving rescue (M3 → M4)
#  c) Social context signals driving rescue (M4 → M5)
#
# Inputs (expected):
#   delta_meds, delta_care, delta_social: pd.DataFrame with columns:
#     - feature_b
#     - smd_A_minus_B
#     - |smd|
#   is_social_feature: function(feature_name:str)->bool (you already have it)
#
# Output:
#   make_fingerprint_1x3_figure(...) returns (fig, axes)
# ============================================================

# -----------------------------
# Color system (match your Fig 1)
# -----------------------------
POS_COLOR = "#E15759"  # positive SMD
NEG_COLOR = "#4C78A8"  # negative SMD

# -----------------------------
# Text helpers
# -----------------------------
def _wrap(s: str, width: int = 30, max_lines: int = 2) -> str:
    """
    Wrap long labels to reduce horizontal space.
    Keeps at most max_lines.
    """
    lines = textwrap.wrap(str(s), width=width)
    if len(lines) <= max_lines:
        return "\n".join(lines)
    return "\n".join(lines[: max_lines - 1] + [" ".join(lines[max_lines - 1 :])])

def _remove_axes_box(ax):
    """Minimal, journal-like axes."""
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_linewidth(0.9)
    ax.spines["bottom"].set_linewidth(0.9)

def _titlecase_value(x: str) -> str:
    """
    Conservative title casing for categorical values.
    Keeps short tokens readable.
    """
    x = str(x).strip()
    if not x:
        return x
    # Title-case but keep acronyms
    parts = x.split()
    out = []
    for p in parts:
        if p.isupper() and len(p) <= 5:
            out.append(p)
        else:
            out.append(p[:1].upper() + p[1:].lower())
    return " ".join(out)

# ============================================================
# Panel A: Medication naming
# ============================================================
# -----------------------------
# MEDS compact naming (robust)
# -----------------------------
_MED_SHORT = {
    "htn_acei": "ACE inhibitor",
    "htn_arb": "ARB",
    "htn_beta_blockers": "Beta blocker",
    "htn_ccb_dhp": "DHP CCB",
    "htn_ccb_non_dhp": "Non-DHP CCB",
    "htn_thiazides": "Thiazide diuretic",
    "htn_loops": "Loop diuretic",
    "htn_potassium_sparing": "Potassium-sparing antihypertensive",
    "metformin": "Metformin",
    "insulin": "Insulin",
    "sulfonylureas": "Sulfonylurea",
    "sglt2i": "SGLT2 inhibitor",
    "glp1ra": "GLP-1 receptor agonist",
    "lipid_statins": "Statin",
    "thiazolidinediones": "Thiazolidinedione",
}

def pretty_meds_compact(f: str, *, wrap: bool = False) -> str:
    """
    Examples:
      htn_acei_recent_30d  -> ACE inhibitor (recent 30 days)
      htn_acei_ever        -> ACE inhibitor (ever used)
      thiazolidinediones_ever -> Thiazolidinedione (ever used)
    """
    # normalize input (protects against accidental whitespace/case issues)
    s = str(f).strip().lower()

    # parse suffix
    if s.endswith("_recent_30d"):
        suffix = " (recent 30 days)"
        base = s[:-len("_recent_30d")]
    elif s.endswith("_ever"):
        suffix = " (ever used)"
        base = s[:-len("_ever")]
    else:
        suffix = ""
        base = s

    # normalize base too (defensive)
    base = base.strip()

    # map
    label = _MED_SHORT.get(base, base.replace("_", " "))

    out = f"{label}{suffix}"
    return _wrap(out) if wrap else out

# ============================================================
# Panel B: Monitoring / labs naming
# ============================================================
_CARE_BASE_RENAMES = {
    "UACR": "UACR",
    "eGFR": "eGFR",
    "FBG": "Fasting glucose",
    "FPG": "Fasting glucose",
    "RBG": "Random glucose",
    "HbA1c": "HbA1c",
    "Hgb": "Hemoglobin",
    "TG": "Triglycerides",
    "HR": "Heart rate",
    "SBP": "SBP",
    "DBP": "DBP",
    "K": "Potassium",
    "HCO3": "Bicarbonate",
    "LDL": "LDL",
    "Weight": "Weight",
    "BUN": "BUN",
    "Screatinine": "Serum creatinine",
    "serum_creatinine": "Serum creatinine",
    "UricAcid": "Uric acid",
    "Na": "Sodium",
    "Cl": "Chloride",
}

def pretty_care_compact(f: str, *, wrap: bool = False) -> str:
    """
    Examples:
      SBP_recency_days -> SBP recency (days)
      eGFR_frac_abnormal -> eGFR fraction abnormal
    """
    s = str(f).strip()

    suffix = ""
    if s.endswith("_recency_days"):
        suffix = " recency (days)"
        base = s[:-len("_recency_days")]
    elif s.endswith("_mean"):
        suffix = " (mean)"
        base = s[:-len("_mean")]
    elif s.endswith("_min"):
        suffix = " (min)"
        base = s[:-len("_min")]
    elif s.endswith("_max"):
        suffix = " (max)"
        base = s[:-len("_max")]
    elif s.endswith("_sd"):
        suffix = " (SD)"
        base = s[:-len("_sd")]
    elif s.endswith("_cov"):
        suffix = " (CV)"
        base = s[:-len("_cov")]
    elif s.endswith("_n"):
        suffix = " (count)"
        base = s[:-len("_n")]
    elif s.endswith("_ever_abnormal"):
        suffix = " ever abnormal"
        base = s[:-len("_ever_abnormal")]
    elif s.endswith("_frac_abnormal"):
        suffix = " fraction abnormal"
        base = s[:-len("_frac_abnormal")]
    elif s.endswith("_slope_dir"):
        suffix = " slope direction"
        base = s[:-len("_slope_dir")]
    elif s.endswith("_slope"):
        suffix = " slope"
        base = s[:-len("_slope")]
    else:
        base = s

    label = _CARE_BASE_RENAMES.get(base, base.replace("_", " "))
    out = f"{label}{suffix}"
    out = re.sub(r"\s+", " ", out).strip()
    return _wrap(out) if wrap else out

# ============================================================
# Panel C: Social naming (including missingness flags)
# ============================================================
_SOCIAL_SPECIAL_COMPACT = {
    "harassment_score_mean": "Everyday discrimination (mean)",
    "harassment_freq_high": "Everyday discrimination (high)",
    "mistrust_mean": "Healthcare mistrust (mean)",
    "mistrust_high": "Healthcare mistrust (high)",
    "stress_mean": "Perceived stress (mean)",
    "stress_high": "Perceived stress (high)",
    "loneliness_mean": "Loneliness (mean)",
    "loneliness_high": "Loneliness (high)",
    "neighborhood_disorder_mean": "Neighborhood disorder (mean)",
}

_SOCIAL_PREFIXES = [
    ("employment_status_", "Employment"),
    ("insurance_status_", "Insurance status"),
    ("insurance_type_", "Insurance type"),
    ("marital_status_", "Marital status"),
    ("living_", "Living"),
    ("race_", "Race"),
]

def pretty_social_compact_panelC(f: str, *, wrap: bool = False) -> str:
    """
    Converts social feature names to compact publication labels.

    Handles:
      - categorical prefixes: employment_status_*, marital_status_*, race_*
      - explicit missingness flags: *_missing and rec_*_missing
      - specific publication-friendly labels
      - fallback cleanup for uncaught variable names
    """
    s = str(f).strip()

    # Robust recreational-drug handling
    if s.startswith("rec_any_drug"):
        out = "Recreational drug use"
        return _wrap(out) if wrap else out

    # Explicit publication-friendly labels
    special_labels = {
        "income_level_missing": "Income level: Missing",
        "housing_unstable_missing": "Housing instability: Missing",
        "education_level_missing": "Education level: Missing",
        "rec_any_drug": "Recreational drug use",
        "employment_status_unable_to_work": "Employment: Unable to work",
    }

    if s in special_labels:
        out = special_labels[s]
        return _wrap(out) if wrap else out

    # Existing hard-coded social labels
    if s in _SOCIAL_SPECIAL_COMPACT:
        out = _SOCIAL_SPECIAL_COMPACT[s]
        return _wrap(out) if wrap else out

    # Explicit missingness indicators
    missing_exact = {
        "home_own_missing": "Missing: home ownership",
        "active_duty_missing": "Missing: active duty status",
    }

    if s in missing_exact:
        out = missing_exact[s]
        return _wrap(out) if wrap else out

    # rec_*_missing patterns
    # Example: rec_any_drug_missing
    if s.startswith("rec_") and s.endswith("_missing"):
        core = s[len("rec_") : -len("_missing")]
        core = core.replace("any_", "").replace("_", " ").strip()

        if core == "drug":
            core = "recreational drug"

        core = core[:1].upper() + core[1:]
        out = f"Missing: {core}"
        return _wrap(out) if wrap else out

    # Continuous / numeric fields
    if s == "income_level":
        out = "Income level"
        return _wrap(out) if wrap else out

    # Prefix-based categorical cleanup
    head = None
    val = s

    for pref, label in _SOCIAL_PREFIXES:
        if s.startswith(pref):
            head = label
            val = s[len(pref):]
            break

    # Fallback for uncaught variable names
    if head is None:
        s2 = re.sub(r"\s+", " ", s.replace("_", " ")).strip()
        s2 = s2[:1].upper() + s2[1:]
        return _wrap(s2) if wrap else s2

    # Normalize categorical value tokens
    val = re.sub(r"\s+", " ", val.replace("_", " ")).strip()

    # Common shortening
    val = val.replace(
        "out of work less than one",
        "out of work <1 year"
    )

    # Race: "any black" -> "Black"
    if head == "Race":
        m = re.match(r"^(any)\s+(.+)$", val, flags=re.I)

        if m:
            val = _titlecase_value(m.group(2))
        else:
            val = _titlecase_value(val)

    else:
        val = _titlecase_value(val)

    out = f"{head}: {val}"
    return _wrap(out) if wrap else out

# ============================================================
# Core plotter: ax-aware fingerprint plot
# ============================================================
def plot_fingerprint_on_ax(
    ax,
    delta_df,
    *,
    top_n=15,
    feature_filter=None,
    feature_prettify=None,
    xlabel="Standardized mean difference\n(newly detected − still missed)",
    show_zero_line=True,
    zero_line_color="0.35",
    zero_line_lw=0.9,
    ytick_fs=8,
    xtick_fs=9,
    xgrid_alpha=0.18,
    bar_edge_lw=0.7,
):
    """
    Draw horizontal bars of standardized mean differences (SMD).
    Positive = more common/higher in Rescued; Negative = more common/higher in Still missed.
    """
    df = delta_df.copy()

    # Optional feature filtering (domain-specific patterns)
    if feature_filter is not None:
        df = df[df["feature_b"].astype(str).apply(feature_filter)].copy()

    if df.empty:
        raise ValueError("No features available after filtering.")

    # Rank by absolute effect size, then order bottom-to-top by signed effect
    df = df.sort_values("|smd|", ascending=False).head(top_n)
    df = df.sort_values("smd_A_minus_B")

    y = np.arange(len(df))
    smd = df["smd_A_minus_B"].to_numpy(dtype=float)

    # Pretty labels
    labels_raw = df["feature_b"].astype(str).to_list()
    labels = [feature_prettify(x) for x in labels_raw] if feature_prettify else labels_raw

    # Color by sign
    colors = np.where(smd > 0, POS_COLOR, NEG_COLOR)

    ax.barh(y, smd, color=colors, edgecolor="black", linewidth=bar_edge_lw)

    # Zero line (lighter / quieter)
    if show_zero_line:
        ax.axvline(0, color=zero_line_color, linewidth=zero_line_lw, zorder=5)

    ax.set_yticks(y)
    ax.set_yticklabels(labels, fontsize=ytick_fs)
    ax.set_xlabel(xlabel, fontsize=10)
    ax.tick_params(axis="x", labelsize=xtick_fs)

    _remove_axes_box(ax)
    ax.grid(axis="x", alpha=xgrid_alpha)

    return df

# ============================================================
# Figure builder: 1x3 combined figure (Figure 2)
# ============================================================
def make_fingerprint_1x3_figure(
    delta_meds,
    delta_care,
    delta_social,
    *,
    top_n_meds=15,
    top_n_care=15,
    top_n_social=15,
    figsize=(18, 5.6),
    dpi=600,
    outpath_base=None,   # stem only
    panel_labels=("a", "b", "c"),
    wrap_labels=False,
    add_panel_titles=True,
    symmetric_xlim=True,
):
    """
    Builds a single 1x3 figure with consistent styling and harmonized x-limits.
    """
    fig, axes = plt.subplots(1, 3, figsize=figsize, dpi=dpi)
    ax1, ax2, ax3 = axes

    # a) Meds (M2 -> M3)
    plot_fingerprint_on_ax(
        ax1, delta_meds,
        top_n=top_n_meds,
        feature_filter=None,
        feature_prettify=(lambda s: pretty_meds_compact(s, wrap=wrap_labels)),
    )

    # b) Care (M3 -> M4) - keep typical lab summary features
    care_mask = lambda s: str(s).endswith((
        "_n", "_recency_days", "_slope", "_slope_dir", "_ever_abnormal", "_frac_abnormal",
        "_mean", "_min", "_max", "_sd", "_cov"
    ))
    plot_fingerprint_on_ax(
        ax2, delta_care,
        top_n=top_n_care,
        feature_filter=care_mask,
        feature_prettify=(lambda s: pretty_care_compact(s, wrap=wrap_labels)),
    )

    # c) Social (M4 -> M5)
    plot_fingerprint_on_ax(
        ax3, delta_social,
        top_n=top_n_social,
        feature_filter=is_social_feature,  # you already have this
        feature_prettify=(lambda s: pretty_social_compact_panelC(s, wrap=wrap_labels)),
    )

    # Small titles per panel (clean, journal-like)
    if add_panel_titles:
        ax1.set_title(
            "Medication signals distinguishing\n"
            "newly detected vs. still-missed cases\n"
            "(M2 → M3)",
            fontsize=10.5, pad=10, loc="left"
        )
    
        ax2.set_title(
            "Kidney trajectory and monitoring signals distinguishing\n"
            "newly detected vs. still-missed cases\n"
            "(M3 → M4)",
            fontsize=10.5, pad=10, loc="left"
        )
    
        ax3.set_title(
            "Social data availability signals distinguishing\n"
            "newly detected vs. still-missed cases\n"
            "(M4 → M5)",
            fontsize=10.5, pad=10, loc="left"
        )
    # Panel letters (larger, bold)
    for lab, ax in zip(panel_labels, axes):
        ax.text(
            -0.08, 1.13, lab,
            transform=ax.transAxes,
            fontsize=16,
            fontweight="bold",
            va="bottom",
            ha="left"
        )

    # Harmonize x-limits across panels (so effect sizes are comparable)
    if symmetric_xlim:
        xmin = min(ax.get_xlim()[0] for ax in axes)
        xmax = max(ax.get_xlim()[1] for ax in axes)
        m = max(abs(xmin), abs(xmax))
        for ax in axes:
            ax.set_xlim(-m, m)
    else:
        xmin = min(ax.get_xlim()[0] for ax in axes)
        xmax = max(ax.get_xlim()[1] for ax in axes)
        for ax in axes:
            ax.set_xlim(xmin, xmax)

    fig.tight_layout(w_pad=4.0)

    if outpath_base is not None:
        fig.savefig(f"{outpath_base}.pdf", bbox_inches="tight")
        fig.savefig(f"{outpath_base}.png", dpi=dpi, bbox_inches="tight")

    return fig, axes
```

# 5. Exploratory M4→M5 rescue by pre-landmark clinical domain coverage

This analysis is retained as an exploratory supplementary analysis rather than a main-figure result.

Here, **domain coverage** is the number of non-kidney general clinical domains with at least one recorded pre-landmark measurement. Because this score is discrete (0–8), nominal deciles collapse to fewer realized quantile groups when ties are removed.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# =============================================================================
# Clinical observability of M4 → M5 rescues (1-year landmark)
#
# Goal
# - Define "observability" as clinical data density in NON-social domains,
#   measured before the 1-year landmark (so it reflects what exists in the record).
# - Make a probability-by-decile curve:
#     y-axis: P(rescued | observability decile)
#     x-axis: observability decile (or median observability in that decile)
#
# Rescued definition
# - Rescued: M4 outcome = FN and M5 outcome = TP
# - Still FN: M4 outcome = FN and M5 outcome = FN
#
# Observability definition (aligned with domain logic)
#   obs_general_any = number of general clinical domains with ≥1 measurement
#   using *_n counts (measurement frequency) as "measurement exists".
#
#
# Important:
# - This code assumes df_y1_m4_features already contains features computed
#   using only data prior to the 1-year landmark (your pipeline already does this).
# =============================================================================


# =============================================================================
# 0) Helper: define which measurement bases count as "general observability"
#    Excludes kidney markers in main figure to avoid "outcome-defining lab" concerns.
#    Put kidney into supplement if desired.
# =============================================================================
GENERAL_DOMAINS = {
    "glycemia":       ["HbA1c", "RBG"],
    "blood_pressure": ["SBP", "DBP"],
    "lipids":         ["LDL", "HDL", "TG", "TotalChol"],
    "electrolytes":   ["Na", "K", "Cl", "HCO3"],
    "liver_protein":  ["ALT", "AST", "ALP", "Albumin", "TotalBilirubin", "DirectBilirubin", "TotalProtein"],
    "hematology":     ["Hgb"],
    "anthropometrics":["Weight"],
    "vitals":         ["HR"],
}

# Supplement option: kidney observability
KIDNEY_BASES = ["eGFR", "Screatinine", "UACR", "UrAlb"]


# =============================================================================
# 1) Build transition labels (rescued vs still-FN) from merged transitions
# =============================================================================
def build_transition_from_merged(
    merged_transitions_df: pd.DataFrame,
    *,
    id_col: str = "person_id",
    a: str = "M4",
    b: str = "M5",
) -> pd.DataFrame:
    """
    Input requires:
      - outcome_{a}, outcome_{b} columns (values like "TP","FP","TN","FN")
      - id_col

    Output:
      - rows restricted to FN→TP ("rescued") and FN→FN ("still FN")
      - columns: [id_col, rescued] where rescued is 1 if FN→TP else 0
    """
    df = merged_transitions_df.copy()

    oa = df[f"outcome_{a}"].astype(str)
    ob = df[f"outcome_{b}"].astype(str)

    rescued = oa.eq("FN") & ob.eq("TP")
    stillfn = oa.eq("FN") & ob.eq("FN")
    keep = rescued | stillfn

    out = df.loc[keep, [id_col]].copy()
    out["rescued"] = rescued.loc[keep].astype(int).values
    return out


# =============================================================================
# 2) Observability construction from *_n columns
# =============================================================================
def _row_sum_numeric(df_subset: pd.DataFrame) -> pd.Series:
    """
    Safe row-wise sum after numeric coercion.
    """
    if df_subset.shape[1] == 0:
        return pd.Series(np.nan, index=df_subset.index)
    x = df_subset.apply(pd.to_numeric, errors="coerce")
    return x.sum(axis=1)


def build_general_observability(
    df_features: pd.DataFrame,
    transition_df: pd.DataFrame,
    *,
    id_col: str = "person_id",
    domains: dict[str, list[str]] | None = None,
    include_kidney: bool = False,
) -> tuple[pd.DataFrame, dict]:
    """
    Build observability metrics aligned to domain framing (excluding social/genetics).

    Observability is based on measurement availability via *_n features.

    Returns:
      obs_df columns:
        - person_id
        - rescued (0/1)
        - obs_general_any    : # domains with ≥1 measurement (domain coverage)
        - obs_general_count  : total measurement burden (sum of *_n across bases)
      audit dict with:
        - which domains/bases were used
        - which *_n columns were found
    """
    domains = domains or GENERAL_DOMAINS
    domains_use = dict(domains)

    if include_kidney:
        domains_use = dict(domains_use)  # copy
        domains_use["kidney"] = list(KIDNEY_BASES)

    # Keep only people relevant for the rescue analysis
    df = df_features.merge(
        transition_df[[id_col, "rescued"]],
        on=id_col,
        how="inner",
    ).copy()

    # For each domain: domain is "observed" if ANY base has *_n > 0
    domain_flags = {}
    used_n_cols = []

    for dom, bases in domains_use.items():
        n_cols = [f"{b}_n" for b in bases if f"{b}_n" in df.columns]

        if len(n_cols) == 0:
            # If a whole domain has no *_n columns, mark as 0 for all rows
            domain_flags[dom] = pd.Series(0, index=df.index, dtype=int)
            continue

        used_n_cols.extend(n_cols)

        x = df[n_cols].apply(pd.to_numeric, errors="coerce").fillna(0.0)
        domain_flags[dom] = (x.sum(axis=1) > 0).astype(int)

    domain_flag_df = pd.DataFrame(domain_flags)

    # Option A: domain coverage score
    df["obs_general_any"] = domain_flag_df.sum(axis=1).astype(int)

    # Option B: total measurement burden
    used_n_cols = list(dict.fromkeys(used_n_cols))  # de-duplicate while preserving order
    df["obs_general_count"] = _row_sum_numeric(df[used_n_cols]).fillna(0.0)

    obs_df = df[[id_col, "rescued", "obs_general_any", "obs_general_count"]].copy()

    audit = {
        "domains_used": domains_use,
        "used_n_cols": used_n_cols,
        "n_domains_possible": int(len(domains_use)),
        "include_kidney": bool(include_kidney),
    }
    return obs_df, audit


# =============================================================================
# 3) Decile summary + Wilson confidence intervals
# =============================================================================
def _wilson_ci(k: float, n: float, z: float = 1.96) -> tuple[float, float]:
    """
    Wilson score interval for a binomial proportion.
    """
    if n <= 0:
        return (np.nan, np.nan)
    p = k / n
    denom = 1 + z**2 / n
    center = (p + z**2 / (2 * n)) / denom
    half = (z * np.sqrt((p * (1 - p) + z**2 / (4 * n)) / n)) / denom
    return (max(0.0, center - half), min(1.0, center + half))


def build_deciles_and_rates(
    obs_df: pd.DataFrame,
    *,
    x_col: str,
    deciles: int = 10,
) -> pd.DataFrame:
    """
    Bin individuals into deciles by observability and compute rescue rate per decile.

    Output columns:
      - obs_decile (1..deciles)
      - n, n_rescued, rescue_rate
      - median_x (median observability in decile, used for plotting x)
      - ci_low, ci_high (Wilson 95% CI)
    """
    x = obs_df.copy()
    x = x.replace([np.inf, -np.inf], np.nan).dropna(subset=[x_col, "rescued"])

    # Handle case where qcut would fail (all values identical)
    if x[x_col].nunique() < 2:
        x["obs_decile"] = 1
    else:
        x["obs_decile"] = pd.qcut(
            x[x_col],
            deciles,
            labels=False,
            duplicates="drop",
        ) + 1

    d = (
        x.groupby("obs_decile")
        .agg(
            n=("rescued", "size"),
            n_rescued=("rescued", "sum"),
            rescue_rate=("rescued", "mean"),
            median_x=(x_col, "median"),
        )
        .reset_index()
        .sort_values("obs_decile")
    )

    d["ci_low"] = [_wilson_ci(k, n)[0] for k, n in zip(d["n_rescued"], d["n"])]
    d["ci_high"] = [_wilson_ci(k, n)[1] for k, n in zip(d["n_rescued"], d["n"])]
    return d


# =============================================================================
# 4) Plot: rescue probability by observability decile (no boxplots)
# =============================================================================
def plot_rescue_probability_by_decile(
    decile_df: pd.DataFrame,
    *,
    x_mode: str = "decile",     # "decile" or "median"
    obs_label: str,
    title: str,
    outpath: str | None = None,
    dpi: int = 600,
):
    """
    x_mode:
      - "decile": x-axis is 1..10 (clean, interpretable)
      - "median": x-axis is median observability within decile (more quantitative)
    """
    d = decile_df.copy().sort_values("obs_decile")

    if x_mode == "decile":
        x = d["obs_decile"].astype(float).values
        xlab = "Clinical domain-coverage quantile group (lowest to highest)"
    elif x_mode == "median":
        x = pd.to_numeric(d["median_x"], errors="coerce").values
        xlab = f"Median {obs_label} within decile"
    else:
        raise ValueError("x_mode must be 'decile' or 'median'")

    y = pd.to_numeric(d["rescue_rate"], errors="coerce").clip(0, 1).values
    lo = pd.to_numeric(d["ci_low"], errors="coerce").values
    hi = pd.to_numeric(d["ci_high"], errors="coerce").values

    plt.rcParams.update({
        "font.size": 10,
        "axes.titlesize": 11,
        "axes.labelsize": 10,
        "xtick.labelsize": 9,
        "ytick.labelsize": 9,
        "axes.spines.top": False,
        "axes.spines.right": False,
    })

    fig = plt.figure(figsize=(5.6, 4.0), dpi=120)
    ax = plt.gca()

    # Confidence band + curve
    ax.fill_between(x, lo, hi, alpha=0.18, linewidth=0)
    ax.plot(x, y, marker="o", linewidth=2.5, markersize=6)

    ax.set_ylim(0, 1)
    ax.set_xlabel(xlab)
    ax.set_ylabel("P(M4 FN → M5 TP | M4 FN)")
    ax.set_title(title)
    ax.grid(alpha=0.25)

    # If using deciles, force integer ticks
    if x_mode == "decile":
        ax.set_xticks(np.arange(1, int(np.nanmax(x)) + 1))

    fig.tight_layout()

    if outpath is not None:
        fig.savefig(outpath + ".png", dpi=dpi, bbox_inches="tight")
        fig.savefig(outpath + ".pdf", bbox_inches="tight")

    return fig, ax


# =============================================================================
# 5) End-to-end runner
# =============================================================================
def run_observability_rescue_curve(
    *,
    merged_transitions_df: pd.DataFrame,
    df_y1_m4_features: pd.DataFrame,
    id_col: str = "person_id",
    a: str = "M4",
    b: str = "M5",
    obs_mode: str = "any",          # "any" or "count"
    include_kidney: bool = False,   # main=False; supplement=True
    deciles: int = 10,
    x_mode: str = "decile",         # "decile" or "median"
    outpath: str | None = None,
    dpi: int = 600,
):
    """
    Produces:
      - obs_df: person-level rescued + observability score
      - decile_df: decile-level rescue rates + CIs
      - fig: probability-by-decile curve
    """
    # Build transition labels (only FN->TP vs FN->FN)
    transition_df = build_transition_from_merged(
        merged_transitions_df,
        id_col=id_col,
        a=a,
        b=b,
    )

    # Build observability (domain coverage or burden)
    obs_df, audit = build_general_observability(
        df_features=df_y1_m4_features,
        transition_df=transition_df,
        id_col=id_col,
        domains=GENERAL_DOMAINS,
        include_kidney=include_kidney,
    )

    x_col = "obs_general_any" if obs_mode == "any" else "obs_general_count"

    # Build deciles + rescue rates
    decile_df = build_deciles_and_rates(obs_df, x_col=x_col, deciles=deciles)

    # Labels for plotting
    if obs_mode == "any":
        obs_label = "domain coverage (count of domains with ≥1 measurement)"
        title = "M4→M5 case rescue by pre-landmark clinical domain coverage"
    else:
        obs_label = "measurement burden (sum of measurement counts)"
        title = "M4→M5 case rescue by pre-landmark measurement density"

    if include_kidney:
        title += " including kidney monitoring"
    else:
        title += " excluding kidney monitoring"

    # Plot curve (no boxplots)
    fig, ax = plot_rescue_probability_by_decile(
        decile_df,
        x_mode=x_mode,
        obs_label=obs_label,
        title=title,
        outpath=outpath,
        dpi=dpi,
    )

    return {
        "transition_df": transition_df,
        "obs_df": obs_df,
        "audit": audit,
        "decile_df": decile_df,
        "fig": fig,
        "ax": ax,
    }
```

```python
def build_domain_map(feature_names, *, drop_empty=True):
    """
    Build a domain map for SHAP aggregation.

    Notes on design choices
    - "monitoring_pattern": biomarker-specific measurement density proxies:
        * *_n (number of measurements)
        * *_recency_days (time since last measurement)
      This is not true utilization (visits/encounters), but a monitoring proxy.

    - Any *_missing flags that are NOT derived from labs/vitals/biomarkers are
      routed to social_access (survey completeness / missingness is often
      access-related).
    """
    # Domain buckets
    demographics, anthropometrics, vitals = ([] for _ in range(3))
    glycemic_labs, non_glycemic_labs, kidney_markers = ([] for _ in range(3))
    monitoring_pattern, abnormality_flags = ([] for _ in range(2))
    comorbidity, treatment, social_access, genetics, family_history = ([] for _ in range(5))
    unmapped = []

    # ----------------------------
    # Sets / prefixes for routing
    # ----------------------------
    comorbidity_set = {
        "hypertension","coronary_artery_disease","myocardial_infarction","heart_failure",
        "atrial_fibrillation","peripheral_arterial_disease","obesity","metabolic_syndrome","gout",
        "acute_kidney_injury","nephrolithiasis","uti","copd","sleep_apnea","steatotic_liver_disease",
        "chronic_liver_disease","chronic_hepatitis_c","cirrhosis","rheumatoid_arthritis","sle","psoriasis",
        "major_depressive_disorder","anxiety_disorder","psychotic_disorder","hiv",
        "diabetic_retinopathy","diabetic_neuropathy","diabetic_foot",
        "tobacco_dependence_syndrome","alcohol_dependence",
    }

    demo_set = {"age_at_t2d", "sex_male", "t2d_duration"}

    anthropo_prefixes = ("Height_", "Weight_", "WHR_")
    anthropo_exact = {"Height", "Weight", "WHR"}

    vital_prefixes = ("SBP_", "DBP_", "HR_")
    vital_base = {"SBP", "DBP", "HR"}

    kidney_prefixes = ("eGFR_", "Screatinine_", "UACR_", "UrAlb_")
    kidney_base = {"eGFR", "Screatinine", "UACR", "UrAlb"}

    glycemic_prefixes = ("HbA1c_", "FBG_", "RBG_")
    glycemic_base = {"HbA1c", "FBG", "RBG"}

    nonglycemic_lab_prefixes = (
        "Na_", "K_", "Cl_", "HCO3_", "Mg_",
        "Hgb_", "ESR_",
        "LDL_", "HDL_", "TG_", "TotalChol_",
        "ALT_", "AST_", "ALP_", "TotalBilirubin_", "DirectBilirubin_",
        "Albumin_", "TotalProtein_",
        "BUN_", "UricAcid_",
    )
    nonglycemic_base = {
        "Na","K","Cl","HCO3","Mg",
        "Hgb","ESR",
        "LDL","HDL","TG","TotalChol",
        "ALT","AST","ALP","TotalBilirubin","DirectBilirubin",
        "Albumin","TotalProtein",
        "BUN","UricAcid",
    }

    # Monitoring intensity proxies (measurement density + recency)
    monitoring_suffixes = ("_n", "_recency_days")

    # Abnormality flags
    abnormality_suffixes = ("_ever_abnormal", "_frac_abnormal")

    # ----------------------------
    # Social: explicit + prefixes
    # ----------------------------
    social_explicit = {
        "housing_unstable",
        "housing_unstable_missing",
        "sex_male_missing",  # survey completeness flag in your data
        "active_duty_yes","active_duty_no","active_duty_missing",
        "has_basic_survey","has_lifestyle_survey","has_sdoh_survey",
    }

    social_prefixes = (
        "education_", "education_level",
        "income_", "income_level",
        "insurance_", "insurance_status_", "insurance_type_",
        "employment_", "employment_status_",
        "living_", "living_situation_",
        "home_", "home_own_",
        "race_", "race_any_",
        "marital_", "marital_status_",
        "hispanic_", "hispanic_any",
        "active_duty_",
        "has_",
        "lonely_", "loneliness_",
        "stress_",
        "mistrust_",
        "harass_", "harassment_",
        "disorder_",
        "neighborhood_",
        "alcohol_",
        "cigar_",
        "cigarette_",
        "cigs_",
        "hookah_",
        "vape_",
        "smokeless_",
        "marijuana_",
        "smoke_",
        "quit_",
        "serious_",
        "rec_",
    )

    def _startswith_any(x: str, prefixes) -> bool:
        return any(x.startswith(p) for p in prefixes)

    def _is_biomarker_derived(x: str) -> bool:
        """
        True if a feature is a lab/vital/biomarker family feature
        (including summary stats like *_mean, *_sd, *_slope, etc).
        Used to route generic *_missing flags into social_access only if
        they are NOT biomarker-derived.
        """
        return (
            _startswith_any(x, vital_prefixes)
            or _startswith_any(x, glycemic_prefixes)
            or _startswith_any(x, kidney_prefixes)
            or _startswith_any(x, nonglycemic_lab_prefixes)
        )

    def uniq(xs):
        seen, out = set(), []
        for x in xs:
            if x not in seen:
                seen.add(x)
                out.append(x)
        return out

    # ----------------------------
    # Routing rules (order matters)
    # ----------------------------
    for f in feature_names:
        if f in {"person_id", "dkd_binary"}:
            continue

        # Genetics + family history
        if (f.startswith("PGS") or "_PRS_" in f or f.lower().startswith("prs_") or "polygenic" in f.lower()):
            genetics.append(f); continue
        if f.startswith("fh_"):
            family_history.append(f); continue

        # Social access / SDoH
        # Route explicit social items, known social prefixes,
        # and generic *_missing flags that are NOT biomarker-derived.
        if (
            f in social_explicit
            or _startswith_any(f, social_prefixes)
            or (f.endswith("_missing") and not _is_biomarker_derived(f))
        ):
            social_access.append(f); continue

        # Monitoring intensity proxies (measurement density + recency)
        if f.endswith(monitoring_suffixes):
            monitoring_pattern.append(f); continue

        # Abnormality flags
        if f.endswith(abnormality_suffixes):
            abnormality_flags.append(f); continue

        # Treatment (med exposures)
        if (
            f.endswith("_ever")
            or f.endswith("_recent_30d")
            or f.startswith("htn_")
            or f.startswith("lipid_")
        ):
            treatment.append(f); continue

        # Comorbidities
        if f in comorbidity_set:
            comorbidity.append(f); continue

        # Demographics
        if f in demo_set:
            demographics.append(f); continue

        # Anthropometrics
        if (f in anthropo_exact) or f.startswith(anthropo_prefixes):
            anthropometrics.append(f); continue

        # Vitals
        if f.startswith(vital_prefixes) or f in vital_base:
            vitals.append(f); continue

        # Kidney markers
        if f.startswith(kidney_prefixes) or f in kidney_base:
            kidney_markers.append(f); continue

        # Glycemic labs
        if f.startswith(glycemic_prefixes) or f in glycemic_base:
            glycemic_labs.append(f); continue

        # Non-glycemic labs
        if f.startswith(nonglycemic_lab_prefixes) or f in nonglycemic_base:
            non_glycemic_labs.append(f); continue

        unmapped.append(f)

    if unmapped:
        raise ValueError(
            "Unmapped features found (should be zero). Add routing rules for:\n"
            + "\n".join(sorted(set(unmapped)))
        )

    domain_map = {
        "demographics": uniq(demographics),
        "anthropometrics": uniq(anthropometrics),
        "vitals": uniq(vitals),
        "glycemic_labs": uniq(glycemic_labs),
        "non_glycemic_labs": uniq(non_glycemic_labs),
        "kidney_markers": uniq(kidney_markers),
        "monitoring_pattern": uniq(monitoring_pattern),
        "abnormality_flags": uniq(abnormality_flags),
        "comorbidity": uniq(comorbidity),
        "family_history": uniq(family_history),
        "treatment": uniq(treatment),
        "social_access": uniq(social_access),
        "genetics": uniq(genetics),
    }

    if drop_empty:
        domain_map = {k: v for k, v in domain_map.items() if len(v) > 0}

    return domain_map
```

```python
DOMAIN_MAP_full = build_domain_map(
    list(X_m5_test.columns)
)

def compute_domain_abs_shap_per_person(
    shap_values,
    feature_names,
    domain_map,
    agg="sum",
):
    shap_abs = np.abs(np.asarray(shap_values))
    feat_idx = {f: i for i, f in enumerate(feature_names)}

    domains = sorted(domain_map.keys())
    mat = np.zeros((shap_abs.shape[0], len(domains)), dtype=float)

    for j, dom in enumerate(domains):
        cols = [c for c in domain_map[dom] if c in feat_idx]
        if not cols:
            continue
        idx = np.array([feat_idx[c] for c in cols], dtype=int)
        if agg == "sum":
            mat[:, j] = shap_abs[:, idx].sum(axis=1)
        elif agg == "mean":
            mat[:, j] = shap_abs[:, idx].mean(axis=1)
        else:
            raise ValueError("agg must be 'sum' or 'mean'")

    dom_abs = pd.DataFrame(mat, columns=domains)
    total = dom_abs.sum(axis=1).replace(0, np.nan)
    dom_frac = dom_abs.div(total, axis=0).fillna(0.0)
    return dom_abs, dom_frac

dom_abs_df, dom_frac_df = compute_domain_abs_shap_per_person(
    shap_values_m5,
    list(X_m5_test.columns),
    DOMAIN_MAP_full,
    agg="sum",
)

domain_summary = pd.DataFrame({
    "domain": dom_abs_df.columns,
    "mean_abs_shap_sum": dom_abs_df.mean(axis=0).values,
    "mean_fraction": dom_frac_df.mean(axis=0).values,
    "n_features": [len(DOMAIN_MAP_full[d]) for d in dom_abs_df.columns],
}).sort_values("mean_fraction", ascending=False)

save_gcs_csv(
    domain_summary,
    f"{RESULT_DIR}/shap_domain_summary_M5_1y_{FINAL_DATE}.csv",
)

display(domain_summary)
```

```python
# ============================================================
# Genetics-dominant SHAP redistribution analysis
# M5, 1-year landmark
# ============================================================

if "genetics" not in dom_frac_df.columns:
    raise KeyError(
        "genetics domain not found. Available domains: "
        + ", ".join(dom_frac_df.columns)
    )

# ------------------------------------------------------------
# Define genetics-dominant participants
# Top 20% of genetics SHAP fraction
# ------------------------------------------------------------
genetics_fraction = dom_frac_df["genetics"].copy()

GENETICS_QUANTILE = 0.80
genetics_cut = float(
    genetics_fraction.quantile(GENETICS_QUANTILE)
)

genetics_dominant = (
    genetics_fraction >= genetics_cut
)


# ============================================================
# 1) Domain-level SHAP redistribution
# ============================================================

# ------------------------------------------------------------
# Point estimates
# ------------------------------------------------------------
genetics_domain_mean = (
    pd.DataFrame({
        "genetics_dominant": genetics_dominant,
    })
    .join(dom_frac_df)
    .groupby("genetics_dominant")
    .mean()
    .T
)

genetics_domain_mean = genetics_domain_mean.rename(
    columns={
        False: "Not genetics-dominant",
        True: "Genetics-dominant",
    }
)

domain_delta_df = (
    genetics_domain_mean["Genetics-dominant"]
    - genetics_domain_mean["Not genetics-dominant"]
).rename(
    "delta_fraction"
).reset_index().rename(
    columns={"index": "domain"}
)


# ------------------------------------------------------------
# Bootstrap 95% CIs for domain redistribution
# Genetics-dominant minus Not genetics-dominant
# ------------------------------------------------------------
rng_domain = np.random.default_rng(SEED)

dom_frac_genetics = (
    dom_frac_df
    .loc[genetics_dominant]
    .copy()
)

dom_frac_other = (
    dom_frac_df
    .loc[~genetics_dominant]
    .copy()
)

boot_domain_delta = np.zeros(
    (N_BOOT, dom_frac_df.shape[1]),
    dtype=float,
)

for b in range(N_BOOT):

    # Stratified bootstrap within each group
    i1 = rng_domain.integers(
        0,
        len(dom_frac_genetics),
        size=len(dom_frac_genetics),
    )

    i0 = rng_domain.integers(
        0,
        len(dom_frac_other),
        size=len(dom_frac_other),
    )

    mean_dom_boot = (
        dom_frac_genetics
        .iloc[i1]
        .mean(axis=0)
        .to_numpy()
    )

    mean_other_boot = (
        dom_frac_other
        .iloc[i0]
        .mean(axis=0)
        .to_numpy()
    )

    boot_domain_delta[b] = (
        mean_dom_boot
        - mean_other_boot
    )


# ------------------------------------------------------------
# Domain-level 95% CIs
# ------------------------------------------------------------
domain_ci_lower = np.percentile(
    boot_domain_delta,
    2.5,
    axis=0,
)

domain_ci_upper = np.percentile(
    boot_domain_delta,
    97.5,
    axis=0,
)

domain_ci_map_lower = dict(
    zip(
        dom_frac_df.columns,
        domain_ci_lower,
    )
)

domain_ci_map_upper = dict(
    zip(
        dom_frac_df.columns,
        domain_ci_upper,
    )
)

domain_delta_df["ci_lower_95"] = (
    domain_delta_df["domain"]
    .map(domain_ci_map_lower)
)

domain_delta_df["ci_upper_95"] = (
    domain_delta_df["domain"]
    .map(domain_ci_map_upper)
)

domain_delta_df["ci_excludes_0"] = (
    (domain_delta_df["ci_lower_95"] > 0)
    |
    (domain_delta_df["ci_upper_95"] < 0)
)

# Manuscript-ready formatted column
domain_delta_df["delta_fraction_95ci"] = (
    domain_delta_df.apply(
        lambda r: (
            f"{r['delta_fraction']:+.3f} "
            f"({r['ci_lower_95']:+.3f} to "
            f"{r['ci_upper_95']:+.3f})"
        ),
        axis=1,
    )
)


# ============================================================
# 2) Individual PRS feature SHAP comparison
# ============================================================

# ------------------------------------------------------------
# Identify PRS features in M5
# ------------------------------------------------------------
prs_features_m5 = [
    c
    for c in X_m5_test.columns
    if (
        c.startswith("PGS")
        or "_PRS_" in c
        or c.lower().startswith("prs_")
        or "polygenic" in c.lower()
    )
]

if not prs_features_m5:
    raise ValueError(
        "No PRS features detected in the M5 model."
    )


# ------------------------------------------------------------
# Locate PRS columns in SHAP matrix
# ------------------------------------------------------------
feature_index = {
    f: i
    for i, f in enumerate(X_m5_test.columns)
}

prs_idx = np.array(
    [
        feature_index[f]
        for f in prs_features_m5
    ],
    dtype=int,
)


# ------------------------------------------------------------
# Point estimates for absolute SHAP values
# ------------------------------------------------------------
shap_abs = np.abs(
    shap_values_m5
)

mean_dom = (
    shap_abs[
        genetics_dominant
    ][:, prs_idx]
    .mean(axis=0)
)

mean_other = (
    shap_abs[
        ~genetics_dominant
    ][:, prs_idx]
    .mean(axis=0)
)

prs_shap_comparison_m5 = pd.DataFrame({
    "prs_feature": prs_features_m5,
    "mean_abs_shap_non_dominant_m5": mean_other,
    "mean_abs_shap_genetics_dominant_m5": mean_dom,
})

prs_shap_comparison_m5[
    "delta_mean_abs_shap_m5"
] = (
    prs_shap_comparison_m5[
        "mean_abs_shap_genetics_dominant_m5"
    ]
    - prs_shap_comparison_m5[
        "mean_abs_shap_non_dominant_m5"
    ]
)

prs_shap_comparison_m5[
    "ratio_mean_abs_shap_m5"
] = (
    prs_shap_comparison_m5[
        "mean_abs_shap_genetics_dominant_m5"
    ]
    /
    (
        prs_shap_comparison_m5[
            "mean_abs_shap_non_dominant_m5"
        ]
        + 1e-12
    )
)


# ------------------------------------------------------------
# Bootstrap 95% CIs for PRS SHAP differences
# ------------------------------------------------------------
# Separate RNG so adding the domain bootstrap does not alter
# the PRS bootstrap sequence/results.
rng_prs = np.random.default_rng(SEED)

dom_arr = (
    shap_abs[
        genetics_dominant
    ][:, prs_idx]
)

other_arr = (
    shap_abs[
        ~genetics_dominant
    ][:, prs_idx]
)

boot_prs = np.zeros(
    (N_BOOT, len(prs_idx)),
    dtype=float,
)

for b in range(N_BOOT):

    i1 = rng_prs.integers(
        0,
        len(dom_arr),
        size=len(dom_arr),
    )

    i0 = rng_prs.integers(
        0,
        len(other_arr),
        size=len(other_arr),
    )

    boot_prs[b] = (
        dom_arr[i1].mean(axis=0)
        - other_arr[i0].mean(axis=0)
    )


# ------------------------------------------------------------
# PRS-level 95% CIs
# ------------------------------------------------------------
prs_shap_comparison_m5[
    "ci_lower_95_delta_m5"
] = np.percentile(
    boot_prs,
    2.5,
    axis=0,
)

prs_shap_comparison_m5[
    "ci_upper_95_delta_m5"
] = np.percentile(
    boot_prs,
    97.5,
    axis=0,
)

prs_shap_comparison_m5[
    "ci_excludes_0_delta_m5"
] = (
    (
        prs_shap_comparison_m5[
            "ci_lower_95_delta_m5"
        ] > 0
    )
    |
    (
        prs_shap_comparison_m5[
            "ci_upper_95_delta_m5"
        ] < 0
    )
)


# ------------------------------------------------------------
# Sort PRS results
# ------------------------------------------------------------
prs_shap_comparison_m5 = (
    prs_shap_comparison_m5
    .sort_values(
        "delta_mean_abs_shap_m5",
        ascending=False,
    )
    .reset_index(drop=True)
)


# ============================================================
# 3) Save outputs
# ============================================================

save_gcs_csv(
    domain_delta_df,
    (
        f"{RESULT_DIR}/"
        f"shap_genetics_domain_redistribution_"
        f"M5_1y_{FINAL_DATE}.csv"
    ),
)

save_gcs_csv(
    prs_shap_comparison_m5,
    (
        f"{RESULT_DIR}/"
        f"shap_prs_comparison_"
        f"M5_1y_{FINAL_DATE}.csv"
    ),
)


# ============================================================
# 4) Summary output
# ============================================================

print(
    "Genetics-dominant definition: "
    "top 20% of genetics SHAP fraction; "
    f"cutpoint={genetics_cut:.6f}; "
    f"n={int(genetics_dominant.sum()):,}; "
    f"bootstrap={N_BOOT:,}"
)

print("\nDomain redistribution:")
display(
    domain_delta_df
    .sort_values(
        "delta_fraction",
        ascending=False,
    )
)

print("\nPRS-level SHAP comparison:")
display(
    prs_shap_comparison_m5
    .head(15)
)
```

# 7. Figure 3 — observation density, social attribution, and decision-curve analysis

Panel A now asks a more direct question than the previous rescue curve: **does mature-model discrimination vary with the amount of pre-landmark clinical information available?**

Observation density is based on the total count of recorded measurements across eight non-kidney general clinical domains. Tertile cut points are learned from the TRAIN partition and then applied unchanged to TEST. This avoids defining subgroups from TEST outcomes or TEST performance.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# =============================================================================
# Figure 3 (1x3 panels, no panel titles)
#   A: AUROC of M4 and M5 across pre-landmark observation-density tertiles
#   B: Social-dominant SHAP redistribution
#   C: Decision curve analysis for M4 vs M5 in the full TEST cohort
#
# Panel A definition
# - Observation density = total pre-landmark measurement count across the same
#   eight NON-KIDNEY general clinical domains used in the observability analysis.
# - Low / Intermediate / High cut points are learned on TRAIN only.
# - The frozen TEST predictions are then evaluated within those groups.
# =============================================================================


# -----------------------------
# Global style helpers
# -----------------------------
def set_pub_rc():
    plt.rcParams.update({
        "font.size": 10,
        "axes.labelsize": 10,
        "xtick.labelsize": 9,
        "ytick.labelsize": 9,
        "legend.fontsize": 9,
        "axes.spines.top": False,
        "axes.spines.right": False,
        "figure.dpi": 120,
        "savefig.dpi": 600,
    })


def _remove_axes_box(ax, lw=0.9):
    ax.spines["left"].set_linewidth(lw)
    ax.spines["bottom"].set_linewidth(lw)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)


def _add_panel_letter(ax, letter, *, x=-0.12, y=1.02, fs=16):
    ax.text(
        x, y, letter,
        transform=ax.transAxes,
        fontsize=fs,
        fontweight="bold",
        va="bottom",
        ha="left",
    )


# =============================================================================
# Panel A data: TRAIN-defined observation-density tertiles
# =============================================================================
def build_observation_density_scores(feature_df: pd.DataFrame):
    """
    Returns person-level pre-landmark observation density.

    obs_general_count:
      total measurement count across non-kidney general clinical domains.

    obs_general_any:
      number of those domains with >=1 measurement.
    """
    out = feature_df[[ID_COL]].copy()

    used_n_cols = []
    domain_flags = {}

    for domain, bases in GENERAL_DOMAINS.items():
        n_cols = [
            f"{base}_n"
            for base in bases
            if f"{base}_n" in feature_df.columns
        ]

        used_n_cols.extend(n_cols)

        if not n_cols:
            domain_flags[domain] = pd.Series(
                0,
                index=feature_df.index,
                dtype=int,
            )
            continue

        x = (
            feature_df[n_cols]
            .apply(pd.to_numeric, errors="coerce")
            .fillna(0)
        )

        domain_flags[domain] = (x.sum(axis=1) > 0).astype(int)

    flags = pd.DataFrame(domain_flags, index=feature_df.index)

    out["obs_general_any"] = flags.sum(axis=1).astype(int)

    used_n_cols = list(dict.fromkeys(used_n_cols))

    out["obs_general_count"] = (
        feature_df[used_n_cols]
        .apply(pd.to_numeric, errors="coerce")
        .fillna(0)
        .sum(axis=1)
        if used_n_cols
        else 0.0
    )

    return out, used_n_cols


def make_train_based_tertiles(values, split):
    """
    Learn observation-density cut points on TRAIN only,
    then apply those cut points to all participants.
    """
    values = pd.to_numeric(values, errors="coerce").fillna(0)
    train_values = values[split.eq("train")]

    q1, q2 = np.quantile(train_values, [1 / 3, 2 / 3])

    if q2 <= q1:
        ranked = train_values.rank(method="average", pct=True)
        q1 = float(
            train_values.loc[
                ranked.sub(1 / 3).abs().idxmin()
            ]
        )
        q2 = float(
            train_values.loc[
                ranked.sub(2 / 3).abs().idxmin()
            ]
        )

    groups = pd.cut(
        values,
        bins=[-np.inf, q1, q2, np.inf],
        labels=["Low", "Intermediate", "High"],
        include_lowest=True,
    )

    return groups, float(q1), float(q2)


def stratified_auc_bootstrap_ci(y, p, *, n_boot=1000, seed=42):
    """
    Stratified bootstrap 95% CI for AUROC.
    Used only for figure uncertainty; no model fitting occurs.
    """
    y = np.asarray(y).astype(int)
    p = np.asarray(p, dtype=float)

    idx0 = np.where(y == 0)[0]
    idx1 = np.where(y == 1)[0]

    if len(idx0) == 0 or len(idx1) == 0:
        return np.nan, np.nan

    rng = np.random.default_rng(seed)
    vals = []

    for _ in range(n_boot):
        b0 = rng.choice(idx0, size=len(idx0), replace=True)
        b1 = rng.choice(idx1, size=len(idx1), replace=True)
        idx = np.concatenate([b0, b1])
        vals.append(roc_auc_score(y[idx], p[idx]))

    return (
        float(np.quantile(vals, 0.025)),
        float(np.quantile(vals, 0.975)),
    )


def build_observation_density_performance(
    *,
    feature_df: pd.DataFrame,
    split_manifest_df: pd.DataFrame,
    predictions_dict: dict,
    model_ids=("M4", "M5"),
    n_boot=1000,
    seed=42,
):
    density, used_cols = build_observation_density_scores(feature_df)

    split_1y = (
        split_manifest_df.loc[
            split_manifest_df["landmark"].eq("1y"),
            [ID_COL, "split"],
        ]
        .copy()
    )

    split_1y[ID_COL] = pd.to_numeric(
        split_1y[ID_COL],
        errors="raise",
    ).astype("int64")

    density[ID_COL] = pd.to_numeric(
        density[ID_COL],
        errors="raise",
    ).astype("int64")

    density = density.merge(
        split_1y,
        on=ID_COL,
        how="inner",
        validate="one_to_one",
    )

    density["observation_density"], q1, q2 = make_train_based_tertiles(
        density["obs_general_count"],
        density["split"],
    )

    density_test = density.loc[
        density["split"].eq("test")
    ].copy()

    rows = []

    level_order = ["Low", "Intermediate", "High"]

    for model_idx, model_id in enumerate(model_ids):
        pred = predictions_dict["1y"][model_id].loc[
            predictions_dict["1y"][model_id]["split"].eq("test"),
            [ID_COL, PRED_OUTCOME_COL, PROB_COL],
        ].copy()

        pred[ID_COL] = pd.to_numeric(
            pred[ID_COL],
            errors="raise",
        ).astype("int64")

        x = density_test.merge(
            pred,
            on=ID_COL,
            how="inner",
            validate="one_to_one",
        )

        for level_idx, level in enumerate(level_order):
            g = x.loc[x["observation_density"].astype(str).eq(level)].copy()

            y = g[PRED_OUTCOME_COL].astype(int).to_numpy()
            p = g[PROB_COL].astype(float).to_numpy()

            auc = roc_auc_score(y, p)
            auprc = average_precision_score(y, p)

            lo, hi = stratified_auc_bootstrap_ci(
                y,
                p,
                n_boot=n_boot,
                seed=seed + 100 * model_idx + level_idx,
            )

            rows.append({
                "model_id": model_id,
                "observation_density": level,
                "n": int(len(g)),
                "cases": int(y.sum()),
                "case_rate": float(y.mean()),
                "AUROC": float(auc),
                "AUROC_ci_low": lo,
                "AUROC_ci_high": hi,
                "AUPRC": float(auprc),
            })

    out = pd.DataFrame(rows)

    return out, {
        "train_q1": q1,
        "train_q2": q2,
        "used_measurement_count_features": used_cols,
    }


def plot_panel_A_observation_density(
    ax,
    density_performance_df: pd.DataFrame,
    *,
    m4_color="#1E293B",
    m5_color="#E76F51",
):
    level_order = ["Low", "Intermediate", "High"]
    x = np.arange(len(level_order))

    for model_id, color, marker in [
        ("M4", m4_color, "o"),
        ("M5", m5_color, "s"),
    ]:
        d = (
            density_performance_df.loc[
                density_performance_df["model_id"].eq(model_id)
            ]
            .set_index("observation_density")
            .loc[level_order]
            .reset_index()
        )

        y = d["AUROC"].to_numpy(float)
        lo = d["AUROC_ci_low"].to_numpy(float)
        hi = d["AUROC_ci_high"].to_numpy(float)

        yerr = np.vstack([y - lo, hi - y])

        ax.errorbar(
            x,
            y,
            yerr=yerr,
            marker=marker,
            markersize=6,
            linewidth=2.1,
            capsize=3,
            label=model_id,
            color=color,
        )

    n_by_level = (
        density_performance_df.loc[
            density_performance_df["model_id"].eq("M5")
        ]
        .set_index("observation_density")
        .loc[level_order, "n"]
        .astype(int)
    )

    xticklabels = [
        f"{level}\n(n={n_by_level.loc[level]:,})"
        for level in level_order
    ]

    ax.set_xticks(x)
    ax.set_xticklabels(xticklabels)
    ax.set_xlabel("Pre-landmark observation density")
    ax.set_ylabel("AUROC (95% CI)")
    ax.set_ylim(0.50, 0.86)
    ax.axhline(0.50, linestyle="--", linewidth=1.0, color="#9A9A9A")
    ax.grid(axis="y", alpha=0.22)
    ax.legend(frameon=False, loc="lower right")
    _remove_axes_box(ax)


# =============================================================================
# Panel B: Social-dominant redistribution
# =============================================================================
def plot_panel_B_social_redistribution(
    ax,
    social_domain_delta_df: pd.DataFrame,
    *,
    top_n: int | None = None,
    pos_color: str = "#2CA02C",
    neg_color: str = "#D62728",
    zero_lw: float = 0.8,
    bar_edge_lw: float = 0.6,
    legend_loc: str = "lower right",
):
    df = social_domain_delta_df.copy()

    if top_n is not None:
        df["abs"] = df["delta_fraction"].abs()
        df = (
            df.sort_values("abs", ascending=False)
            .head(top_n)
            .drop(columns=["abs"])
        )

    df = df.sort_values("delta_fraction")

    def pretty_domain(s: str) -> str:
        s = str(s).replace("_", " ").strip()
        return (s[:1].upper() + s[1:]) if s else s

    y_labels = [pretty_domain(d) for d in df["domain"]]
    x_vals = pd.to_numeric(
        df["delta_fraction"],
        errors="coerce",
    ).values

    colors = np.where(x_vals >= 0, pos_color, neg_color)

    ax.barh(
        y_labels,
        x_vals,
        color=colors,
        edgecolor="black",
        linewidth=bar_edge_lw,
        height=0.72,
    )

    ax.axvline(0, color="black", linewidth=zero_lw)

    ax.set_xlabel(
        "Difference in mean relative domain contribution (|SHAP|)\n"
        "(Social-dominant minus others)"
    )
    ax.set_ylabel("")
    ax.grid(axis="x", alpha=0.25)
    ax.set_axisbelow(True)
    _remove_axes_box(ax)

    legend_elements = [
        Patch(
            facecolor=pos_color,
            edgecolor="none",
            label="Higher contribution",
        ),
        Patch(
            facecolor=neg_color,
            edgecolor="none",
            label="Lower contribution",
        ),
    ]
    ax.legend(
        handles=legend_elements,
        frameon=False,
        loc=legend_loc,
    )


# =============================================================================
# Panel C: Decision Curve Analysis
# =============================================================================
def _net_benefit(y_true, y_prob, thresholds):
    y_true = np.asarray(y_true).astype(int)
    y_prob = np.asarray(y_prob).astype(float)

    mask = np.isfinite(y_prob)
    y_true = y_true[mask]
    y_prob = y_prob[mask]

    N = y_true.size
    if N == 0:
        return np.full_like(
            thresholds,
            np.nan,
            dtype=float,
        )

    nb = np.empty_like(thresholds, dtype=float)

    for i, pt in enumerate(thresholds):
        pred_pos = y_prob >= pt
        TP = np.sum(pred_pos & (y_true == 1))
        FP = np.sum(pred_pos & (y_true == 0))
        w = pt / (1.0 - pt)
        nb[i] = (TP / N) - (FP / N) * w

    return nb


def _net_benefit_treat_all(y_true, thresholds):
    y_true = np.asarray(y_true).astype(int)

    if y_true.size == 0:
        return np.full_like(
            thresholds,
            np.nan,
            dtype=float,
        )

    prev = np.mean(y_true)
    w = thresholds / (1.0 - thresholds)

    return prev - (1.0 - prev) * w


def plot_panel_C_dca(
    ax,
    merged_preds: pd.DataFrame,
    *,
    y_col: str = "y_true",
    prob_m4_col: str = "prob_m4",
    prob_m5_col: str = "prob_m5",
    t_min: float = 0.01,
    t_max: float = 0.30,
    n_thresholds: int = 250,
    m4_color: str = "#1E293B",
    m5_color: str = "#E76F51",
    ref_color: str = "#7F7F7F",
):
    df = merged_preds.copy()

    y = df[y_col].astype(int).values
    p4 = df[prob_m4_col].astype(float).values
    p5 = df[prob_m5_col].astype(float).values

    thresholds = np.linspace(
        t_min,
        t_max,
        n_thresholds,
    )
    thresholds = np.clip(
        thresholds,
        1e-6,
        1 - 1e-6,
    )

    nb_m4 = _net_benefit(y, p4, thresholds)
    nb_m5 = _net_benefit(y, p5, thresholds)
    nb_all = _net_benefit_treat_all(y, thresholds)
    nb_none = np.zeros_like(thresholds)

    ax.plot(
        thresholds,
        nb_m4,
        linewidth=2.2,
        label="M4",
        color=m4_color,
    )
    ax.plot(
        thresholds,
        nb_m5,
        linewidth=2.2,
        label="M5",
        color=m5_color,
    )
    ax.plot(
        thresholds,
        nb_all,
        linestyle="--",
        linewidth=1.8,
        label="Treat all",
        color=ref_color,
    )
    ax.plot(
        thresholds,
        nb_none,
        linestyle="--",
        linewidth=1.8,
        label="Treat none",
        color=ref_color,
        alpha=0.75,
    )

    ax.set_xlabel("Threshold probability")
    ax.set_ylabel("Net benefit")
    ax.grid(alpha=0.25)
    _remove_axes_box(ax)
    ax.legend(frameon=False, loc="best")


# =============================================================================
# Figure 3 wrapper
# =============================================================================
def make_figure3(
    *,
    density_performance_df: pd.DataFrame,
    social_domain_delta_df: pd.DataFrame,
    merged_preds: pd.DataFrame,
    B_top_n: int | None = None,
    C_t_min: float = 0.01,
    C_t_max: float = 0.30,
    figsize=(15.2, 4.6),
    dpi=600,
    outpath_base: str | None = None,
):
    set_pub_rc()

    fig, axes = plt.subplots(
        1,
        3,
        figsize=figsize,
        dpi=120,
    )

    axA, axB, axC = axes

    plot_panel_A_observation_density(
        axA,
        density_performance_df=density_performance_df,
    )
    _add_panel_letter(axA, "a", fs=16)

    plot_panel_B_social_redistribution(
        axB,
        social_domain_delta_df=social_domain_delta_df,
        top_n=B_top_n,
        pos_color="#2CA02C",
        neg_color="#D62728",
    )
    _add_panel_letter(axB, "b", fs=16)

    plot_panel_C_dca(
        axC,
        merged_preds=merged_preds,
        t_min=C_t_min,
        t_max=C_t_max,
        m4_color="#1E293B",
        m5_color="#E76F51",
        ref_color="#7F7F7F",
    )
    _add_panel_letter(axC, "c", fs=16)

    fig.tight_layout(w_pad=2.0)

    if outpath_base is not None:
        fig.savefig(
            f"{outpath_base}.png",
            dpi=dpi,
            bbox_inches="tight",
        )
        fig.savefig(
            f"{outpath_base}.pdf",
            bbox_inches="tight",
        )

    return fig, axes
```

```python
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch


# =============================================================================
# Global plotting style
# =============================================================================

def set_pub_rc():
    plt.rcParams.update({
        "font.size": 10,
        "axes.labelsize": 10,
        "xtick.labelsize": 9,
        "ytick.labelsize": 9,
        "legend.fontsize": 9,
        "axes.spines.top": False,
        "axes.spines.right": False,
        "figure.dpi": 150,
        "savefig.dpi": 600,
    })


def clean_axes(ax):
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    ax.grid(
        axis="x",
        alpha=0.25,
        linewidth=0.8,
    )

    ax.set_axisbelow(True)


def add_panel_letter(ax, letter):
    ax.text(
        -0.12,
        1.02,
        letter,
        transform=ax.transAxes,
        fontsize=16,
        fontweight="bold",
        ha="left",
        va="bottom",
    )


# =============================================================================
# Panel A — Domain substitution when genetics dominates
# =============================================================================

def plot_genetic_domain_substitution(
    ax,
    domain_delta_df,
):
    df = (
        domain_delta_df
        .copy()
        .sort_values("delta_fraction")
    )

    def pretty_domain(s):
        s = str(s).replace("_", " ").strip()

        if not s:
            return s

        return s[:1].upper() + s[1:]

    y_labels = [
        pretty_domain(d)
        for d in df["domain"]
    ]

    x_vals = (
        df["delta_fraction"]
        .to_numpy()
    )

    colors = np.where(
        x_vals >= 0,
        "#2ca02c",
        "#d62728",
    )

    ax.barh(
        y_labels,
        x_vals,
        color=colors,
        edgecolor="black",
        linewidth=0.6,
        height=0.72,
    )

    ax.axvline(
        0,
        color="black",
        linewidth=0.8,
    )

    ax.set_xlabel(
        "Difference in mean relative domain contribution (|SHAP|)\n"
        "(Genetics-dominant minus others)"
    )

    ax.set_ylabel("")

    clean_axes(ax)

    legend_elements = [
        Patch(
            facecolor="#2ca02c",
            edgecolor="black",
            label="Higher contribution",
        ),
        Patch(
            facecolor="#d62728",
            edgecolor="black",
            label="Lower contribution",
        ),
    ]

    ax.legend(
        handles=legend_elements,
        frameon=False,
        loc="lower right",
    )


# =============================================================================
# Panel B — Individual PRS effects
# =============================================================================

PRS_NAME_MAP = {
    "egfr": "eGFR",
    "bmi": "BMI",
    "homair": "HOMA-IR",
    "homa_ir": "HOMA-IR",
    "homa": "HOMA",
    "ckd": "CKD",
    "hypertension": "Hypertension",
    "uacr": "UACR",
    "dbp": "DBP",
    "sbp": "SBP",
    "ldl": "LDL",
    "hdl": "HDL",
    "triglycerides": "Triglycerides",
    "smoking": "Smoking",
    "type2diabetes": "T2D",
    "t2d": "T2D",
    "microalbuminuria": "Microalbuminuria",
    "albuminuria": "Albuminuria",
    "creatinine": "Creatinine",
    "diabetes": "Diabetes",
}


def pretty_prs_label(s):
    """
    Examples
    --------
    eGFR_PRS_PGS000884
        -> eGFR PRS (PGS000884)

    BMI_PRS_PGS005198
        -> BMI PRS (PGS005198)

    Type2Diabetes_PRS_PGS000036
        -> T2D PRS (PGS000036)

    HOMAIR_PRS_PGS003470
        -> HOMA-IR PRS (PGS003470)
    """

    s = str(s).strip()

    # ---------------------------------------------------------
    # Extract PGS identifier wherever it occurs
    # ---------------------------------------------------------
    pgs_match = re.search(
        r"PGS[_\s-]*(\d+)",
        s,
        flags=re.IGNORECASE,
    )

    pgs = None

    if pgs_match:
        pgs = f"PGS{pgs_match.group(1).zfill(6)}"

        # Remove PGS token from phenotype portion
        s = (
            s[:pgs_match.start()]
            + s[pgs_match.end():]
        )

    # ---------------------------------------------------------
    # Remove PRS token
    # ---------------------------------------------------------
    s = re.sub(
        r"(?i)(^|[_\s-])PRS($|[_\s-])",
        " ",
        s,
    )

    # Normalize separators
    s = re.sub(
        r"[_\-]+",
        " ",
        s,
    )

    s = re.sub(
        r"\s+",
        " ",
        s,
    ).strip()

    # ---------------------------------------------------------
    # First try an exact normalized phenotype match
    # ---------------------------------------------------------
    normalized = re.sub(
        r"[^A-Za-z0-9]",
        "",
        s,
    ).lower()

    if normalized in PRS_NAME_MAP:
        label = PRS_NAME_MAP[normalized]

    else:
        # -----------------------------------------------------
        # Generic fallback:
        # split camel case without destroying acronyms
        # -----------------------------------------------------
        label = re.sub(
            r"(?<=[a-z])(?=[A-Z])",
            " ",
            s,
        )

        label = re.sub(
            r"(?<=[A-Za-z])(?=\d)",
            " ",
            label,
        )

        label = re.sub(
            r"(?<=\d)(?=[A-Za-z])",
            " ",
            label,
        )

        label = re.sub(
            r"\s+",
            " ",
            label,
        ).strip()

        # Apply known acronym substitutions token-wise
        words = []

        for word in label.split():
            key = re.sub(
                r"[^A-Za-z0-9]",
                "",
                word,
            ).lower()

            if key in PRS_NAME_MAP:
                words.append(
                    PRS_NAME_MAP[key]
                )
            else:
                words.append(
                    word.capitalize()
                )

        label = " ".join(words)

    # ---------------------------------------------------------
    # Final standardized label
    # ---------------------------------------------------------
    if pgs:
        return f"{label} PRS ({pgs})"

    return f"{label} PRS"


def plot_prs_importance(
    ax,
    prs_shap_comparison_m5,
    top_k=12,
):
    df = (
        prs_shap_comparison_m5
        .sort_values(
            "delta_mean_abs_shap_m5",
            ascending=False,
        )
        .head(top_k)
        .copy()
    )

    y_labels = [
        pretty_prs_label(s)
        for s in df["prs_feature"]
    ]

    x_vals = (
        df["delta_mean_abs_shap_m5"]
        .to_numpy()
    )

    # Reverse so largest appears at top
    y_pos = np.arange(
        len(df)
    )

    ax.barh(
        y_pos,
        x_vals[::-1],
        color="#1f4e8c",
        edgecolor="black",
        linewidth=0.6,
        height=0.72,
    )

    ax.set_yticks(
        y_pos
    )

    ax.set_yticklabels(
        y_labels[::-1],
        fontsize=9,
    )

    # Slight gap between labels and y-axis
    ax.tick_params(
        axis="y",
        pad=5,
    )

    ax.axvline(
        0,
        color="black",
        linewidth=0.8,
    )

    ax.set_xlabel(
        "Increase in mean |SHAP| in genetics-dominant group"
    )

    ax.set_ylabel("")

    clean_axes(ax)


# =============================================================================
# Figure 4 wrapper
# =============================================================================

def make_figure4(
    domain_delta_df,
    prs_shap_comparison_m5,
    *,
    top_k=12,
    figsize=(12.5, 4.8),
    outpath=None,
    dpi=600,
):
    set_pub_rc()

    fig, axes = plt.subplots(
        1,
        2,
        figsize=figsize,
        gridspec_kw={
            "width_ratios": [1.0, 1.15],
        },
    )

    axA, axB = axes

    # -------------------------------------------------------------------------
    # Panel A
    # -------------------------------------------------------------------------
    plot_genetic_domain_substitution(
        axA,
        domain_delta_df,
    )

    add_panel_letter(
        axA,
        "A",
    )

    # -------------------------------------------------------------------------
    # Panel B
    # -------------------------------------------------------------------------
    plot_prs_importance(
        axB,
        prs_shap_comparison_m5,
        top_k=top_k,
    )

    add_panel_letter(
        axB,
        "B",
    )

    # More room between panels for Panel B's longer y-axis labels
    fig.subplots_adjust(
        wspace=0.48,
    )

    plt.tight_layout()

    if outpath is not None:
        fig.savefig(
            f"{outpath}.png",
            bbox_inches="tight",
            dpi=dpi,
        )

        fig.savefig(
            f"{outpath}.pdf",
            bbox_inches="tight",
        )

        fig.savefig(
            f"{outpath}.tiff",
            bbox_inches="tight",
            dpi=dpi,
        )

    return fig, axes
```

```python
def dominance_sensitivity(domain_name, quantiles=(0.90, 0.80, 0.70)):
    frac = dom_frac_df[domain_name]
    rows = []

    for q in quantiles:
        cut = float(frac.quantile(q))
        flag = frac >= cut
        means1 = dom_frac_df.loc[flag].mean()
        means0 = dom_frac_df.loc[~flag].mean()
        delta = means1 - means0

        for domain in dom_frac_df.columns:
            rows.append({
                "dominant_domain": domain_name,
                "top_percent": int(round((1-q) * 100)),
                "quantile": q,
                "cutpoint": cut,
                "n_dominant": int(flag.sum()),
                "domain": domain,
                "delta_fraction": float(delta[domain]),
            })

    return pd.DataFrame(rows)

dominance_sensitivity_df = pd.concat(
    [
        dominance_sensitivity("genetics"),
        dominance_sensitivity("social_access"),
    ],
    ignore_index=True,
)

save_gcs_csv(
    dominance_sensitivity_df,
    f"{RESULT_DIR}/shap_dominance_cutoff_sensitivity_10_20_30_{FINAL_DATE}.csv",
)

display(
    dominance_sensitivity_df.loc[
        dominance_sensitivity_df["domain"].isin(["genetics", "social_access"])
    ]
)
```

# 10. Supplementary Figure S1 — ROC and PR across all corrected landmarks


```python
from __future__ import annotations

from pathlib import Path
from typing import Optional, Sequence

import matplotlib.pyplot as plt
import matplotlib.lines as mlines
import numpy as np
from sklearn.metrics import (
    average_precision_score,
    precision_recall_curve,
    roc_auc_score,
    roc_curve,
)

# ---------------------------------------------------------------------
# Style
# ---------------------------------------------------------------------
PALETTE = {
    "M1":  "#6E6E6E",
    "M2":  "#3B82F6",
    "M3":  "#10B981",
    "M4":  "#1E293B",
    "M4a": "#1E293B",
    "M5":  "#E76F51",
}

CURVE_STYLE = {
    "M1":  dict(ls="--", lw=1.8, alpha=0.95),
    "M2":  dict(ls="-",  lw=1.8, alpha=0.95),
    "M3":  dict(ls="-",  lw=1.8, alpha=0.95),
    "M4":  dict(ls="-",  lw=1.9, alpha=0.95),
    "M4a": dict(ls="-",  lw=1.9, alpha=0.95),
    "M5":  dict(ls="-",  lw=2.4, alpha=1.00),
}


def _display_model_name(k: str) -> str:
    """
    Publication-facing model name.

    Keeps the internal M4a key unchanged so p_M4a and aligned["M4a"]
    continue to work, but displays M4 in the figure.
    """
    return "M4" if k == "M4a" else k


def _set_pub_rc() -> None:
    plt.rcParams.update(
        {
            "font.family": "sans-serif",
            "font.size": 10,
            "axes.titlesize": 12,
            "axes.labelsize": 10.5,
            "xtick.labelsize": 9,
            "ytick.labelsize": 9,
            "legend.fontsize": 9,
            "axes.linewidth": 0.9,
            "savefig.bbox": "tight",
        }
    )


def _remove_spines(ax: plt.Axes) -> None:
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_linewidth(0.9)
    ax.spines["bottom"].set_linewidth(0.9)


def _soft_grid(ax: plt.Axes) -> None:
    ax.grid(
        True,
        alpha=0.10,
        linewidth=0.7,
    )


def _metric_box(
    ax: plt.Axes,
    lines: list[str],
    *,
    loc: str = "upper left",
    fontsize: float = 7.4,
) -> None:
    """
    Draw a compact text-only metric box inside a panel.
    """

    xy = {
        "upper left":  (0.03, 0.97),
        "upper right": (0.97, 0.97),
        "lower left":  (0.03, 0.03),
        "lower right": (0.97, 0.03),
    }[loc]

    ha = "left" if "left" in loc else "right"
    va = "top" if "upper" in loc else "bottom"

    ax.text(
        xy[0],
        xy[1],
        "\n".join(lines),
        transform=ax.transAxes,
        ha=ha,
        va=va,
        fontsize=fontsize,
        color="#222222",
        bbox=dict(
            boxstyle="round,pad=0.25",
            facecolor="white",
            edgecolor="#D9D9D9",
            linewidth=0.7,
            alpha=0.92,
        ),
    )


def _compute_panel_metrics(
    aligned: dict,
    order: list[str],
) -> tuple[list[str], list[str], float]:

    roc_lines = []
    pr_lines = []

    y0 = (
        aligned[order[0]]["y_true"]
        .to_numpy()
        .astype(int)
    )

    prevalence = float(
        np.mean(y0)
    )

    for k in order:
        df = aligned[k]

        y = (
            df["y_true"]
            .to_numpy()
            .astype(int)
        )

        p = (
            df[f"p_{k}"]
            .to_numpy()
            .astype(float)
        )

        auc = roc_auc_score(
            y,
            p,
        )

        ap = average_precision_score(
            y,
            p,
        )

        display_k = _display_model_name(k)

        roc_lines.append(
            f"{display_k}: {auc:.3f}"
        )

        pr_lines.append(
            f"{display_k}: {ap:.3f}"
        )

    return (
        roc_lines,
        pr_lines,
        prevalence,
    )


def _plot_roc(
    ax: plt.Axes,
    aligned: dict,
    order: list[str],
    *,
    xpad: float = 0.02,
) -> None:

    ax.plot(
        [0, 1],
        [0, 1],
        ls=":",
        lw=0.9,
        color="#BBBBBB",
        zorder=1,
    )

    for k in order:

        df = aligned[k]

        y = (
            df["y_true"]
            .to_numpy()
            .astype(int)
        )

        p = (
            df[f"p_{k}"]
            .to_numpy()
            .astype(float)
        )

        fpr, tpr, _ = roc_curve(
            y,
            p,
        )

        style = CURVE_STYLE.get(
            k,
            dict(
                ls="-",
                lw=1.8,
                alpha=0.95,
            ),
        )

        ax.plot(
            fpr,
            tpr,
            color=PALETTE.get(
                k,
                "black",
            ),
            **style,
        )

    ax.set_xlim(
        -xpad,
        1.0,
    )

    ax.set_ylim(
        0.0,
        1.0,
    )

    ax.set_xlabel(
        "False Positive Rate"
    )

    ax.set_ylabel(
        "True Positive Rate"
    )

    _remove_spines(ax)
    _soft_grid(ax)


def _plot_pr(
    ax: plt.Axes,
    aligned: dict,
    order: list[str],
    *,
    pr_ymin: float = 0.30,
) -> float:

    y0 = (
        aligned[order[0]]["y_true"]
        .to_numpy()
        .astype(int)
    )

    prev = float(
        np.mean(y0)
    )

    ax.axhline(
        prev,
        ls=":",
        lw=1.0,
        color="#BFBFBF",
        zorder=1,
    )

    for k in order:

        df = aligned[k]

        y = (
            df["y_true"]
            .to_numpy()
            .astype(int)
        )

        p = (
            df[f"p_{k}"]
            .to_numpy()
            .astype(float)
        )

        prec, rec, _ = precision_recall_curve(
            y,
            p,
        )

        style = CURVE_STYLE.get(
            k,
            dict(
                ls="-",
                lw=1.8,
                alpha=0.95,
            ),
        )

        ax.plot(
            rec,
            prec,
            color=PALETTE.get(
                k,
                "black",
            ),
            **style,
        )

    ax.set_xlim(
        0.0,
        1.0,
    )

    ax.set_ylim(
        pr_ymin,
        1.0,
    )

    ax.set_xlabel(
        "Recall"
    )

    ax.set_ylabel(
        "Precision"
    )

    _remove_spines(ax)
    _soft_grid(ax)

    return prev


def plot_all_landmarks_roc_pr_clean(
    landmark_results: Sequence[dict],
    landmark_labels: Sequence[str],
    *,
    pr_ymin: float = 0.30,
    roc_xpad: float = 0.02,
    output_path: Optional[str] = None,
    dpi: int = 300,
    show_fig: bool = True,
) -> tuple[plt.Figure, np.ndarray]:

    if len(landmark_results) != len(landmark_labels):
        raise ValueError(
            "`landmark_results` and `landmark_labels` "
            "must have the same length."
        )

    _set_pub_rc()

    n_landmarks = len(
        landmark_results
    )

    fig_w = max(
        6.5,
        3.3 * n_landmarks,
    )

    fig_h = 7.0

    fig, axes = plt.subplots(
        nrows=2,
        ncols=n_landmarks,
        figsize=(
            fig_w,
            fig_h,
        ),
        gridspec_kw={
            "hspace": 0.34,
            "wspace": 0.24,
        },
    )

    if n_landmarks == 1:
        axes = np.array(
            [
                [axes[0]],
                [axes[1]],
            ]
        )

    # ---------------------------------------------------------
    # Shared model legend
    # ---------------------------------------------------------
    first_aligned = (
        landmark_results[0]["aligned"]
    )

    model_order = list(
        first_aligned.keys()
    )

    handles = []

    for k in model_order:

        handles.append(
            mlines.Line2D(
                [],
                [],
                color=PALETTE.get(
                    k,
                    "black",
                ),
                linestyle=CURVE_STYLE.get(
                    k,
                    {},
                ).get(
                    "ls",
                    "-",
                ),
                linewidth=CURVE_STYLE.get(
                    k,
                    {},
                ).get(
                    "lw",
                    1.8,
                ),
                label=_display_model_name(k),
            )
        )

    # ---------------------------------------------------------
    # Plot landmarks
    # ---------------------------------------------------------
    for col, (res, label) in enumerate(
        zip(
            landmark_results,
            landmark_labels,
        )
    ):

        aligned = res["aligned"]

        order = list(
            aligned.keys()
        )

        ax_roc = axes[
            0,
            col,
        ]

        ax_pr = axes[
            1,
            col,
        ]

        label_roc = chr(
            ord("a")
            + col * 2
        )

        label_pr = chr(
            ord("a")
            + col * 2
            + 1
        )

        # -----------------------------------------------------
        # Panel labels
        # -----------------------------------------------------
        ax_roc.text(
            -0.09,
            1.12,
            label_roc,
            transform=ax_roc.transAxes,
            fontsize=12,
            fontweight="bold",
            va="top",
            ha="right",
            clip_on=False,
        )

        ax_pr.text(
            -0.09,
            1.12,
            label_pr,
            transform=ax_pr.transAxes,
            fontsize=12,
            fontweight="bold",
            va="top",
            ha="right",
            clip_on=False,
        )

        # -----------------------------------------------------
        # ROC
        # -----------------------------------------------------
        _plot_roc(
            ax_roc,
            aligned,
            order,
            xpad=roc_xpad,
        )

        ax_roc.set_title(
            label,
            fontsize=12,
            fontweight="semibold",
            pad=6,
        )

        # -----------------------------------------------------
        # Precision-recall
        # -----------------------------------------------------
        _plot_pr(
            ax_pr,
            aligned,
            order,
            pr_ymin=pr_ymin,
        )

        # -----------------------------------------------------
        # Metrics
        # -----------------------------------------------------
        roc_lines, pr_lines, _ = (
            _compute_panel_metrics(
                aligned,
                order,
            )
        )

        _metric_box(
            ax_roc,
            ["ROCAUC"]
            + roc_lines,
            loc="lower right",
            fontsize=7.2,
        )

        _metric_box(
            ax_pr,
            ["AUPRC"]
            + pr_lines,
            loc="lower left",
            fontsize=7.2,
        )

        # -----------------------------------------------------
        # Only keep y-axis labels on first column
        # -----------------------------------------------------
        if col != 0:

            ax_roc.set_ylabel("")
            ax_pr.set_ylabel("")

            ax_roc.tick_params(
                axis="y",
                labelleft=False,
            )

            ax_pr.tick_params(
                axis="y",
                labelleft=False,
            )

    # ---------------------------------------------------------
    # Row labels
    # ---------------------------------------------------------
    fig.text(
        0.008,
        0.74,
        "ROC",
        va="center",
        ha="left",
        rotation=90,
        fontsize=10,
        color="#444444",
        fontweight="medium",
    )

    fig.text(
        0.008,
        0.26,
        "Precision-Recall",
        va="center",
        ha="left",
        rotation=90,
        fontsize=10,
        color="#444444",
        fontweight="medium",
    )

    # ---------------------------------------------------------
    # Shared legend
    # Push farther down so it does not overlap Recall labels
    # ---------------------------------------------------------
    fig.legend(
        handles=handles,
        labels=[
            h.get_label()
            for h in handles
        ],
        ncol=len(handles),
        loc="lower center",
        bbox_to_anchor=(
            0.5,
            -0.015,
        ),
        frameon=False,
        handlelength=2.0,
        columnspacing=1.4,
        handletextpad=0.5,
    )

    # ---------------------------------------------------------
    # Figure margins
    # Extra bottom space reserved for legend
    # ---------------------------------------------------------
    plt.subplots_adjust(
        top=0.90,
        left=0.075,
        right=0.99,
        bottom=0.15,
    )

    # ---------------------------------------------------------
    # Save
    # ---------------------------------------------------------
    if output_path is not None:

        p = Path(
            output_path
        )

        p.parent.mkdir(
            parents=True,
            exist_ok=True,
        )

        fig.savefig(
            str(
                p.with_suffix(
                    ".pdf"
                )
            )
        )

        fig.savefig(
            str(
                p.with_suffix(
                    ".png"
                )
            ),
            dpi=dpi,
        )

        print(
            f"Saved -> "
            f"{p.with_suffix('.pdf')} | "
            f"{p.with_suffix('.png')}"
        )

    if show_fig:
        plt.show()

    return (
        fig,
        axes,
    )
```

# 11. Supplementary Figure S2 — DCA across all corrected landmarks


```python
from __future__ import annotations
from pathlib import Path
from typing import Optional, Sequence

import matplotlib.pyplot as plt
import matplotlib.lines as mlines
import numpy as np

# ---------------------------------------------------------------------------
# Style constants
# ---------------------------------------------------------------------------
PALETTE = {
    "M1":  "#6E6E6E",
    "M2":  "#3B82F6",
    "M3":  "#10B981",
    "M4":  "#1E293B",
    "M4a": "#1E293B",
    "M5":  "#E76F51",
}

CURVE_STYLE = {
    "M1":  dict(ls="--", lw=2.0, alpha=0.98),
    "M2":  dict(ls="-",  lw=2.1, alpha=0.98),
    "M3":  dict(ls="-",  lw=2.1, alpha=0.98),
    "M4":  dict(ls="-",  lw=2.2, alpha=0.98),
    "M4a": dict(ls="-",  lw=2.2, alpha=0.98),
    "M5":  dict(ls="-",  lw=2.7, alpha=1.00),
}

DISPLAY_NAMES = {
    "M1": "M1",
    "M2": "M2",
    "M3": "M3",
    "M4": "M4",
    "M4a": "M4",
    "M5": "M5",
}


def _set_pub_rc() -> None:
    plt.rcParams.update({
        "font.family": "sans-serif",
        "font.size": 10.5,
        "axes.titlesize": 12,
        "axes.labelsize": 10.5,
        "legend.fontsize": 9.2,
        "xtick.labelsize": 9.3,
        "ytick.labelsize": 9.3,
        "axes.linewidth": 0.9,
        "lines.linewidth": 1.8,
        "savefig.bbox": "tight",
    })


def _style_ax(ax: plt.Axes) -> None:
    for spine in ax.spines.values():
        spine.set_visible(True)
        spine.set_linewidth(0.85)
        spine.set_edgecolor("#8A8A8A")
    ax.grid(True, alpha=0.08, linewidth=0.65)


# ---------------------------------------------------------------------------
# DCA computation
# ---------------------------------------------------------------------------

def _net_benefit(y_true: np.ndarray, y_prob: np.ndarray,
                 thresholds: np.ndarray) -> np.ndarray:
    """
    Net benefit at each threshold probability pt:
      NB(pt) = TP/N - FP/N * pt/(1-pt)
    """
    n = len(y_true)
    nb = np.full_like(thresholds, np.nan, dtype=float)
    for i, pt in enumerate(thresholds):
        if pt >= 1.0:
            nb[i] = np.nan
            continue
        yhat = (y_prob >= pt).astype(int)
        tp = int(((y_true == 1) & (yhat == 1)).sum())
        fp = int(((y_true == 0) & (yhat == 1)).sum())
        nb[i] = tp / n - fp / n * pt / (1.0 - pt)
    return nb


def _nb_treat_all(y_true: np.ndarray, thresholds: np.ndarray) -> np.ndarray:
    """Net benefit of treating everyone."""
    prev = float(np.mean(y_true))
    nb = np.full_like(thresholds, np.nan, dtype=float)
    for i, pt in enumerate(thresholds):
        if pt >= 1.0:
            nb[i] = np.nan
            continue
        nb[i] = prev - (1.0 - prev) * pt / (1.0 - pt)
    return nb


# ---------------------------------------------------------------------------
# Per-panel DCA plot
# ---------------------------------------------------------------------------

def _plot_dca(
    ax: plt.Axes,
    aligned: dict,
    order: list,
    *,
    pt_lo: float = 0.01,
    pt_hi: float = 0.50,
    n_pts: int = 200,
    clip_lo: float = -0.05,
) -> None:
    thresholds = np.linspace(pt_lo, pt_hi, n_pts)

    # Reference y_true
    y_ref = aligned[order[0]]["y_true"].to_numpy().astype(int)

    # Treat-none
    ax.axhline(
        0.0,
        ls=":",
        lw=1.3,
        color="#8F8F8F",
        zorder=1,
        label="Treat none",
    )

    # Treat-all
    nb_all = _nb_treat_all(y_ref, thresholds)
    nb_all = np.clip(nb_all, clip_lo, None)
    ax.plot(
        thresholds,
        nb_all,
        ls="-.",
        lw=1.8,
        color="#9B9B9B",
        alpha=0.95,
        zorder=2,
        label="Treat all",
    )

    # Model curves
    for k in order:
        df = aligned[k]
        y = df["y_true"].to_numpy().astype(int)
        p = df[f"p_{k}"].to_numpy().astype(float)
        nb = _net_benefit(y, p, thresholds)
        nb = np.clip(nb, clip_lo, None)

        st = CURVE_STYLE.get(k, dict(ls="-", lw=2.0, alpha=0.98))
        ax.plot(
            thresholds,
            nb,
            color=PALETTE.get(k, "black"),
            label=DISPLAY_NAMES.get(k, k),
            **st,
        )

    ax.set_xlim(pt_lo, pt_hi)
    ax.set_ylim(clip_lo, None)
    ax.set_xlabel("Threshold probability")
    _style_ax(ax)


# ---------------------------------------------------------------------------
# Main public function
# ---------------------------------------------------------------------------

def plot_all_landmarks_dca(
    landmark_results: Sequence[dict],
    landmark_labels: Sequence[str],
    *,
    pt_lo: float = 0.01,
    pt_hi: float = 0.50,
    n_pts: int = 200,
    clip_lo: float = -0.05,
    output_path: Optional[str] = None,
    dpi: int = 300,
    show_fig: bool = True,
) -> tuple:
    """
    One DCA panel per landmark in a single row.
    """
    if len(landmark_results) != len(landmark_labels):
        raise ValueError("`landmark_results` and `landmark_labels` must have the same length.")

    _set_pub_rc()

    n = len(landmark_results)
    fig_w = max(4.8, 3.45 * n)
    fig_h = 3.9

    fig, axes = plt.subplots(
        nrows=1,
        ncols=n,
        figsize=(fig_w, fig_h),
        gridspec_kw={"wspace": 0.18},
    )

    if n == 1:
        axes = np.array([axes])

    first_aligned = landmark_results[0]["aligned"]
    model_order = list(first_aligned.keys())

    # Shared legend handles
    handles = [
        mlines.Line2D([], [], color="#8F8F8F", linestyle=":",  linewidth=1.3, label="Treat none"),
        mlines.Line2D([], [], color="#9B9B9B", linestyle="-.", linewidth=1.8, label="Treat all"),
    ]

    for k in model_order:
        handles.append(
            mlines.Line2D(
                [],
                [],
                color=PALETTE.get(k, "black"),
                linestyle=CURVE_STYLE.get(k, {}).get("ls", "-"),
                linewidth=CURVE_STYLE.get(k, {}).get("lw", 2.0),
                label=DISPLAY_NAMES.get(k, k),
            )
        )

    for col, (res, lm_label) in enumerate(zip(landmark_results, landmark_labels)):
        aligned = res["aligned"]
        order = list(aligned.keys())
        ax = axes[col]

        _plot_dca(
            ax,
            aligned,
            order,
            pt_lo=pt_lo,
            pt_hi=pt_hi,
            n_pts=n_pts,
            clip_lo=clip_lo,
        )

        ax.set_title(lm_label, fontsize=12, fontweight="semibold", pad=6)

        if col == 0:
            ax.set_ylabel("Net benefit", labelpad=3)
        else:
            ax.set_ylabel("")
            ax.tick_params(axis="y", labelleft=False, length=0)

        letter = chr(ord("a") + col)
        ax.text(
            -0.03,
            1.05,
            letter,
            transform=ax.transAxes,
            fontsize=11.5,
            fontweight="bold",
            va="bottom",
            ha="left",
            clip_on=False,
        )

    # Shared bottom legend
    fig.legend(
        handles=handles,
        labels=[h.get_label() for h in handles],
        ncol=len(handles),
        loc="lower center",
        bbox_to_anchor=(0.5, -0.1),
        frameon=False,
        handlelength=2.2,
        columnspacing=1.2,
        handletextpad=0.45,
    )

    plt.subplots_adjust(
        top=0.88,
        bottom=0.18,
        left=0.075,
        right=0.99,
    )

    if output_path is not None:
        p = Path(output_path)
        p.parent.mkdir(parents=True, exist_ok=True)
        fig.savefig(str(p.with_suffix(".pdf")))
        fig.savefig(str(p.with_suffix(".png")), dpi=dpi)
        print(f"Saved -> {p.with_suffix('.pdf')}  |  {p.with_suffix('.png')}")

    if show_fig:
        plt.show()

    return fig, axes
```

```python
from __future__ import annotations

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import TwoSlopeNorm

# ============================================================
# Supplementary Figure S3
# Observability-stratified case-control prevalence gaps
# AoU + BioMe
# ============================================================


# ------------------------------------------------------------
# 1) Feature definitions
# ------------------------------------------------------------

FEATURE_SPECS = [
    ("Hypertension", [
        "hypertension",
    ]),
    ("CAD", [
        "coronary_artery_disease",
    ]),
    ("Heart failure", [
        "heart_failure",
    ]),
    ("Atrial fibrillation", [
        "atrial_fibrillation",
    ]),
    ("Obesity", [
        "obesity",
    ]),
    ("AKI history", [
        "acute_kidney_injury",
    ]),
    ("Metformin exposure", [
        "metformin_ever",
    ]),
    ("Insulin exposure", [
        "insulin_ever",
    ]),
    ("ACE inhibitor exposure", [
        "htn_acei_ever",
    ]),
    ("Statin exposure", [
        "lipid_statins_ever",
    ]),
    ("Sleep apnea", [
        "sleep_apnea",
    ]),
    ("Major depressive disorder", [
        "major_depressive_disorder",
    ]),
    ("Diabetic retinopathy", [
        "diabetic_retinopathy",
    ]),
    ("Diabetic neuropathy", [
        "diabetic_neuropathy",
    ]),
]

LANDMARKS_FIG = ["30d", "6m", "1y", "2y"]

LANDMARK_TITLES_FIG = {
    "30d": "Baseline/30d landmark",
    "6m": "6m landmark",
    "1y": "1y landmark",
    "2y": "2y landmark",
}


# ------------------------------------------------------------
# 2) Helpers
# ------------------------------------------------------------

def _binary01(s):
    return (
        pd.to_numeric(s, errors="coerce")
        .fillna(0)
        .gt(0)
        .astype(int)
    )


def _resolve_feature(df, candidates):
    """
    Return the first matching feature column.
    Matching is first exact, then case-insensitive.
    """
    for c in candidates:
        if c in df.columns:
            return c

    lower_map = {
        str(c).lower(): c
        for c in df.columns
    }

    for c in candidates:
        if c.lower() in lower_map:
            return lower_map[c.lower()]

    return None


def diag_observability_one_landmark(
    df,
    landmark,
    feature_specs,
):
    """
    Compute case minus control prevalence gap within
    observability tertiles for one landmark.
    """

    n_cols = [
        c
        for c in df.columns
        if str(c).endswith("_n")
    ]

    if not n_cols:
        raise ValueError(
            f"{landmark}: no *_n monitoring features available."
        )

    # Overall observability score
    obs = (
        df[n_cols]
        .apply(pd.to_numeric, errors="coerce")
        .fillna(0)
        .sum(axis=1)
    )

    work = pd.DataFrame(
        {
            OUTCOME_COL: _binary01(df[OUTCOME_COL]),
            "obs": obs,
        },
        index=df.index,
    )

    # Rank first to guarantee 3 similarly sized groups
    ranked = work["obs"].rank(
        method="first"
    )

    work["obs_tertile"] = pd.qcut(
        ranked,
        q=3,
        labels=[1, 2, 3],
        duplicates="drop",
    ).astype(int)

    rows = []

    for pretty_name, candidates in feature_specs:

        feat = _resolve_feature(
            df,
            candidates,
        )

        if feat is None:
            print(
                f"WARNING: {landmark}: "
                f"feature not found for '{pretty_name}'"
            )
            continue

        v = _binary01(
            df[feat]
        )

        for tert in [1, 2, 3]:

            for outcome in [0, 1]:

                mask = (
                    work["obs_tertile"].eq(tert)
                    &
                    work[OUTCOME_COL].eq(outcome)
                )

                n = int(
                    mask.sum()
                )

                prev = (
                    100
                    * float(
                        v.loc[mask].mean()
                    )
                    if n
                    else np.nan
                )

                rows.append(
                    {
                        "landmark": landmark,
                        "feature": pretty_name,
                        "feature_raw": feat,
                        "obs_tertile": tert,
                        "dkd_binary": outcome,
                        "n": n,
                        "prevalence_pct": prev,
                    }
                )

    long_df = pd.DataFrame(
        rows
    )

    out = (
        long_df
        .pivot_table(
            index=[
                "landmark",
                "feature",
                "obs_tertile",
            ],
            columns="dkd_binary",
            values="prevalence_pct",
        )
        .reset_index()
    )

    out.columns.name = None

    out = out.rename(
        columns={
            0: "control_pct",
            1: "case_pct",
        }
    )

    for col in [
        "control_pct",
        "case_pct",
    ]:
        if col not in out.columns:
            out[col] = np.nan

    out[
        "case_minus_control_pct"
    ] = (
        out["case_pct"]
        - out["control_pct"]
    )

    return out


# ------------------------------------------------------------
# 3) Compute AoU diagnostic
# ------------------------------------------------------------

aou_diag_frames = []

for lm in LANDMARKS_FIG:

    aou_diag_frames.append(
        diag_observability_one_landmark(
            MODEL_DF[lm]["M5"],
            lm,
            FEATURE_SPECS,
        )
    )

aou_diag_df = pd.concat(
    aou_diag_frames,
    ignore_index=True,
)

aou_diag_df["cohort"] = "AoU"


# Save AoU data
save_gcs_csv(
    aou_diag_df,
    (
        f"{RESULT_DIR}/"
        f"supp_observability_case_control_gap_AoU_"
        f"{FINAL_DATE}.csv"
    ),
)


# ------------------------------------------------------------
# 4) Load BioMe diagnostic
# ------------------------------------------------------------

BIOME_GCS = (
    "gs://rw-migration-aou-rw-639fade3/"
    "aou_dkd/dkd_biome_results/"
    "biome_external_suppfig3_case_control_gaps_20260823.csv"
)

biome_diag_df = pd.read_csv(
    BIOME_GCS
)

print(
    "BioMe columns:",
    biome_diag_df.columns.tolist(),
)

display(
    biome_diag_df.head()
)


# ------------------------------------------------------------
# 5) Standardize BioMe column names
# ------------------------------------------------------------

# Accept a few possible naming variants
rename_candidates = {
    "case_control_gap_pp": "case_minus_control_pct",
    "case_minus_control_pp": "case_minus_control_pct",
    "gap_pp": "case_minus_control_pct",
    "tertile": "obs_tertile",
    "observability_tertile": "obs_tertile",
}

for old, new in rename_candidates.items():
    if (
        old in biome_diag_df.columns
        and new not in biome_diag_df.columns
    ):
        biome_diag_df = (
            biome_diag_df.rename(
                columns={
                    old: new
                }
            )
        )

required_cols = [
    "landmark",
    "feature",
    "obs_tertile",
    "case_minus_control_pct",
]

missing_cols = [
    c
    for c in required_cols
    if c not in biome_diag_df.columns
]

if missing_cols:
    raise ValueError(
        "BioMe diagnostic file is missing: "
        + ", ".join(missing_cols)
    )

biome_diag_df["cohort"] = "BioMe"


# ------------------------------------------------------------
# 6) Normalize landmark names
# ------------------------------------------------------------

LANDMARK_NORMALIZE = {
    "30d": "30d",
    "30 days": "30d",
    "30-day": "30d",
    "30day": "30d",

    "6m": "6m",
    "6 months": "6m",
    "6-month": "6m",

    "1y": "1y",
    "1 year": "1y",
    "1-year": "1y",

    "2y": "2y",
    "2 years": "2y",
    "2-year": "2y",
}


def normalize_landmark(x):
    x = str(x).strip().lower()

    return LANDMARK_NORMALIZE.get(
        x,
        x,
    )


aou_diag_df["landmark"] = (
    aou_diag_df["landmark"]
    .map(normalize_landmark)
)

biome_diag_df["landmark"] = (
    biome_diag_df["landmark"]
    .map(normalize_landmark)
)


# ------------------------------------------------------------
# 7) Normalize feature labels
# ------------------------------------------------------------

FEATURE_LABEL_MAP = {
    "hypertension": "Hypertension",
    "coronary_artery_disease": "CAD",
    "heart_failure": "Heart failure",
    "atrial_fibrillation": "Atrial fibrillation",
    "obesity": "Obesity",
    "acute_kidney_injury": "AKI history",
    "metformin_ever": "Metformin exposure",
    "insulin_ever": "Insulin exposure",
    "htn_acei_ever": "ACE inhibitor exposure",
    "lipid_statins_ever": "Statin exposure",
    "sleep_apnea": "Sleep apnea",
    "major_depressive_disorder": "Major depressive disorder",
    "diabetic_retinopathy": "Diabetic retinopathy",
    "diabetic_neuropathy": "Diabetic neuropathy",
}

def normalize_feature_name(x):

    s = str(x).strip()

    # Already pretty
    pretty_values = [
        x[0]
        for x in FEATURE_SPECS
    ]

    if s in pretty_values:
        return s

    return FEATURE_LABEL_MAP.get(
        s.lower(),
        s,
    )


aou_diag_df["feature"] = (
    aou_diag_df["feature"]
    .map(normalize_feature_name)
)

biome_diag_df["feature"] = (
    biome_diag_df["feature"]
    .map(normalize_feature_name)
)


# ------------------------------------------------------------
# 8) Combine
# ------------------------------------------------------------

combined_diag_df = pd.concat(
    [
        aou_diag_df[
            [
                "landmark",
                "feature",
                "obs_tertile",
                "case_minus_control_pct",
                "cohort",
            ]
        ],
        biome_diag_df[
            [
                "landmark",
                "feature",
                "obs_tertile",
                "case_minus_control_pct",
                "cohort",
            ]
        ],
    ],
    ignore_index=True,
)


FEATURE_ORDER = [
    x[0]
    for x in FEATURE_SPECS
]


# ------------------------------------------------------------
# 9) Build heatmap matrices
# ------------------------------------------------------------

def make_panel_matrix(
    df,
    landmark,
):

    sub = df.loc[
        df["landmark"].eq(landmark)
    ].copy()

    matrix = np.full(
        (
            len(FEATURE_ORDER),
            6,
        ),
        np.nan,
    )

    # Column order:
    # AoU Low / Mid / High
    # BioMe Low / Mid / High
    cohort_offset = {
        "AoU": 0,
        "BioMe": 3,
    }

    feature_index = {
        f: i
        for i, f in enumerate(
            FEATURE_ORDER
        )
    }

    for _, row in sub.iterrows():

        feat = row["feature"]
        cohort = row["cohort"]

        if (
            feat not in feature_index
            or cohort not in cohort_offset
        ):
            continue

        tert = int(
            row["obs_tertile"]
        )

        if tert not in [1, 2, 3]:
            continue

        ri = feature_index[feat]

        ci = (
            cohort_offset[cohort]
            + tert
            - 1
        )

        matrix[
            ri,
            ci,
        ] = float(
            row[
                "case_minus_control_pct"
            ]
        )

    return matrix


panel_matrices = {
    lm: make_panel_matrix(
        combined_diag_df,
        lm,
    )
    for lm in LANDMARKS_FIG
}


# ------------------------------------------------------------
# 10) Shared color scale
# ------------------------------------------------------------

all_values = np.concatenate(
    [
        m[np.isfinite(m)]
        for m in panel_matrices.values()
    ]
)

max_abs = float(
    np.nanmax(
        np.abs(all_values)
    )
)

# Round outward for a clean symmetric scale
vlim = max(
    5,
    np.ceil(
        max_abs / 2
    ) * 2,
)

norm = TwoSlopeNorm(
    vmin=-vlim,
    vcenter=0,
    vmax=vlim,
)


# ------------------------------------------------------------
# 11) Plot
# ------------------------------------------------------------

plt.rcParams.update(
    {
        "font.family": "sans-serif",
        "font.size": 8.5,
        "axes.titlesize": 10,
        "axes.labelsize": 9,
        "xtick.labelsize": 7.5,
        "ytick.labelsize": 7.5,
        "savefig.dpi": 600,
    }
)

fig, axes = plt.subplots(
    1,
    4,
    figsize=(13.8, 5.1),
    sharey=True,
)

fig.subplots_adjust(
    left=0.16,
    right=0.925,
    bottom=0.18,
    top=0.91,
    wspace=0.16,
)


images = []

for panel_i, (
    ax,
    lm,
) in enumerate(
    zip(
        axes,
        LANDMARKS_FIG,
    )
):

    matrix = panel_matrices[lm]

    im = ax.imshow(
        matrix,
        cmap="RdBu_r",
        norm=norm,
        aspect="auto",
        interpolation="nearest",
    )

    images.append(im)

    # ---------------------------------------------
    # Title
    # ---------------------------------------------
    ax.set_title(
        LANDMARK_TITLES_FIG[lm],
        fontsize=10,
        pad=7,
    )

    # ---------------------------------------------
    # X tick labels
    # ---------------------------------------------
    ax.set_xticks(
        np.arange(6)
    )

    ax.set_xticklabels(
        [
            "Low",
            "Mid",
            "High",
            "Low",
            "Mid",
            "High",
        ]
    )

    # ---------------------------------------------
    # Y labels only on first panel
    # ---------------------------------------------
    ax.set_yticks(
        np.arange(
            len(FEATURE_ORDER)
        )
    )

    if panel_i == 0:
        ax.set_yticklabels(
            FEATURE_ORDER
        )
    else:
        ax.tick_params(
            axis="y",
            labelleft=False,
        )

    # ---------------------------------------------
    # Cohort labels below x-axis
    # ---------------------------------------------
    ax.text(
        1,
        len(FEATURE_ORDER) + 0.45,
        "AoU",
        ha="center",
        va="top",
        fontsize=8.5,
        fontweight="bold",
        clip_on=False,
    )

    ax.text(
        4,
        len(FEATURE_ORDER) + 0.45,
        "BioMe",
        ha="center",
        va="top",
        fontsize=8.5,
        fontweight="bold",
        clip_on=False,
    )

    # divider between cohorts
    ax.axvline(
        2.5,
        color="white",
        linewidth=2.0,
    )

    # ---------------------------------------------
    # Cell grid
    # ---------------------------------------------
    ax.set_xticks(
        np.arange(-0.5, 6, 1),
        minor=True,
    )

    ax.set_yticks(
        np.arange(
            -0.5,
            len(FEATURE_ORDER),
            1,
        ),
        minor=True,
    )

    ax.grid(
        which="minor",
        color="white",
        linewidth=0.8,
    )

    ax.tick_params(
        which="minor",
        bottom=False,
        left=False,
    )

    # ---------------------------------------------
    # Cell annotations
    # ---------------------------------------------
    for i in range(
        matrix.shape[0]
    ):

        for j in range(
            matrix.shape[1]
        ):

            value = matrix[i, j]

            if not np.isfinite(value):
                continue

            # White text on darker cells
            text_color = (
                "white"
                if abs(value) >= 0.55 * vlim
                else "#222222"
            )

            ax.text(
                j,
                i,
                f"{value:+.1f}",
                ha="center",
                va="center",
                fontsize=6.5,
                color=text_color,
            )

    # ---------------------------------------------
    # Borders
    # ---------------------------------------------
    for spine in ax.spines.values():
        spine.set_visible(False)


# ------------------------------------------------------------
# 12) Shared labels
# ------------------------------------------------------------

fig.text(
    0.54,
    0.065,
    "Observability tertile",
    ha="center",
    va="center",
    fontsize=9.5,
)


# ------------------------------------------------------------
# 13) Colorbar
# ------------------------------------------------------------

cbar_ax = fig.add_axes(
    [
        0.94,
        0.22,
        0.014,
        0.58,
    ]
)

cbar = fig.colorbar(
    images[0],
    cax=cbar_ax,
)

cbar.set_label(
    "Case minus control prevalence gap (percentage points)",
    fontsize=8.5,
    rotation=270,
    labelpad=16,
)

cbar.ax.tick_params(
    labelsize=7.5
)


# ------------------------------------------------------------
# 14) Save
# ------------------------------------------------------------

save_figure_all(
    fig,
    "Supplementary_Figure_S3_observability_case_control_gap_AoU_BioMe",
    dpi=600,
)

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter

# ============================================================
# Supplementary Figure S4
# M4 → M5 case recovery by non-kidney clinical domain coverage
#
# Uses existing notebook objects:
#   res_1y, MODEL_DF, RESULT_DIR, FINAL_DATE
#   save_gcs_csv(), save_figure_all()
#
# Existing transition labels already use frozen CAL thresholds.
# No model fitting or threshold selection is performed here.
# ============================================================

S4_DOMAINS = {
    "glycemia": ["HbA1c", "RBG"],
    "blood_pressure": ["SBP", "DBP"],
    "lipids": ["LDL", "HDL", "TG", "TotalChol"],
    "electrolytes": ["Na", "K", "Cl", "HCO3"],
    "liver_protein": [
        "ALT", "AST", "ALP", "Albumin",
        "TotalBilirubin", "DirectBilirubin", "TotalProtein",
    ],
    "hematology": ["Hgb"],
    "anthropometrics": ["Weight"],
    "vitals": ["HR"],
}

# ------------------------------------------------------------
# 1. Extract cases missed by M4
# ------------------------------------------------------------

transitions = res_1y["merged_transitions_df"].copy()
features = MODEL_DF["1y"]["M4"].copy()

required = ["person_id", "outcome_M4", "outcome_M5"]
missing = [c for c in required if c not in transitions.columns]
if missing:
    raise ValueError(f"Transition table is missing: {missing}")

for name, frame in [
    ("Transition table", transitions),
    ("M4 feature table", features),
]:
    ids = pd.to_numeric(frame["person_id"], errors="raise")
    if ids.isna().any() or (ids % 1 != 0).any():
        raise ValueError(f"{name}: missing or non-integer participant IDs.")

    frame["person_id"] = ids.astype("int64")

    if frame["person_id"].duplicated().any():
        raise ValueError(f"{name}: duplicate participant IDs.")

missed = transitions.loc[
    transitions["outcome_M4"].eq("FN"),
    ["person_id", "outcome_M5"],
].copy()

if missed.empty:
    raise ValueError("No M4 false-negative cases were found.")

if not missed["outcome_M5"].isin(["FN", "TP"]).all():
    raise ValueError(
        "M4 false-negative cases must be classified as FN or TP by M5."
    )

missed["rescued"] = missed["outcome_M5"].eq("TP").astype(int)

# ------------------------------------------------------------
# 2. Calculate domain coverage from pre-landmark counts
# ------------------------------------------------------------

coverage = features[["person_id"]].copy()
domain_columns = {}

for domain, bases in S4_DOMAINS.items():
    cols = [
        f"{base}_n"
        for base in bases
        if f"{base}_n" in features.columns
    ]

    if not cols:
        raise ValueError(
            f"No measurement-count columns found for domain '{domain}'."
        )

    counts = features[cols].apply(pd.to_numeric, errors="raise")

    if np.isinf(counts.to_numpy(dtype=float)).any():
        raise ValueError(f"Infinite measurement counts in '{domain}'.")

    if counts.lt(0).any().any():
        raise ValueError(f"Negative measurement counts in '{domain}'.")

    # Same convention as the notebook:
    # missing counts are treated as no recorded measurements.
    coverage[f"observed_{domain}"] = (
        counts.fillna(0).gt(0).any(axis=1).astype(int)
    )
    domain_columns[domain] = cols

flag_cols = [f"observed_{domain}" for domain in S4_DOMAINS]

coverage["domain_coverage"] = (
    coverage[flag_cols].sum(axis=1).astype(int)
)

obs_s4 = missed[["person_id", "rescued"]].merge(
    coverage,
    on="person_id",
    how="left",
    validate="one_to_one",
    indicator=True,
)

if not obs_s4["_merge"].eq("both").all():
    raise ValueError(
        "Some M4 false-negative cases are missing from the feature table."
    )

obs_s4 = obs_s4.drop(columns="_merge")

# ------------------------------------------------------------
# 3. Preserve the original quantile-group analysis
#
# Coverage is discrete, so duplicate decile boundaries are
# removed. These are realized quantile groups, not ten deciles.
# ------------------------------------------------------------

if obs_s4["domain_coverage"].nunique() == 1:
    obs_s4["coverage_group"] = 1
else:
    group_codes = pd.qcut(
        obs_s4["domain_coverage"],
        q=10,
        labels=False,
        duplicates="drop",
    )

    if group_codes.isna().any():
        raise ValueError("Could not assign all cases to coverage groups.")

    # Consecutive labels for occupied groups, preserving their order.
    occupied = sorted(group_codes.unique())
    group_map = {code: i + 1 for i, code in enumerate(occupied)}

    obs_s4["coverage_group"] = (
        group_codes.map(group_map).astype(int)
    )

summary_s4 = (
    obs_s4.groupby("coverage_group", sort=True)
    .agg(
        n=("rescued", "size"),
        n_rescued=("rescued", "sum"),
        coverage_min=("domain_coverage", "min"),
        coverage_median=("domain_coverage", "median"),
        coverage_max=("domain_coverage", "max"),
    )
    .reset_index()
)

summary_s4["rescue_rate"] = (
    summary_s4["n_rescued"] / summary_s4["n"]
)

# Wilson 95% confidence intervals.
z = 1.959963984540054
n = summary_s4["n"].to_numpy(dtype=float)
p = summary_s4["rescue_rate"].to_numpy(dtype=float)

denominator = 1 + z**2 / n
center = (p + z**2 / (2 * n)) / denominator
half_width = (
    z * np.sqrt(p * (1 - p) / n + z**2 / (4 * n**2))
    / denominator
)

summary_s4["ci_low"] = np.clip(center - half_width, 0, 1)
summary_s4["ci_high"] = np.clip(center + half_width, 0, 1)

assert int(summary_s4["n"].sum()) == len(missed)
assert int(summary_s4["n_rescued"].sum()) == int(missed["rescued"].sum())

# ------------------------------------------------------------
# 4. Plot
# ------------------------------------------------------------

plt.rcParams.update({
    "font.family": "sans-serif",
    "font.size": 10,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.linewidth": 0.8,
    "xtick.major.width": 0.8,
    "ytick.major.width": 0.8,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
    "savefig.dpi": 600,
})

x = summary_s4["coverage_group"].to_numpy()
y = summary_s4["rescue_rate"].to_numpy()
lo = summary_s4["ci_low"].to_numpy()
hi = summary_s4["ci_high"].to_numpy()

n_total = int(summary_s4["n"].sum())
n_rescued = int(summary_s4["n_rescued"].sum())
overall_rate = n_rescued / n_total

fig_s4, ax = plt.subplots(figsize=(7.2, 4.6))
fig_s4.subplots_adjust(
    left=0.13,
    right=0.98,
    bottom=0.25,
    top=0.88,
)

ax.errorbar(
    x,
    y,
    yerr=np.maximum(0, np.vstack([y - lo, hi - y])),
    fmt="o-",
    color="#4C78A8",
    linewidth=1.8,
    markersize=6,
    markeredgecolor="white",
    markeredgewidth=0.8,
    elinewidth=1.3,
    capsize=3,
    label="Group estimate (95% CI)",
    zorder=3,
)

ax.axhline(
    overall_rate,
    color="0.45",
    linestyle="--",
    linewidth=1.1,
    label=f"Overall recovery: {overall_rate:.1%}",
)

tick_labels = []
for row in summary_s4.itertuples(index=False):
    low = int(row.coverage_min)
    high = int(row.coverage_max)
    coverage_label = str(low) if low == high else f"{low}–{high}"

    tick_labels.append(
        f"{int(row.coverage_group)}\n"
        f"{coverage_label} domains\n"
        f"(n={int(row.n):,})"
    )

ax.set_xticks(x)
ax.set_xticklabels(tick_labels, fontsize=8.5)
ax.set_xlim(x.min() - 0.4, x.max() + 0.4)

ymax = min(
    1.0,
    max(0.20, float(np.ceil((hi.max() + 0.02) / 0.05) * 0.05)),
)
ax.set_ylim(0, ymax)
ax.yaxis.set_major_formatter(PercentFormatter(xmax=1, decimals=0))

ax.set_xlabel(
    "Pre-landmark domain-coverage quantile group",
    labelpad=9,
)
ax.set_ylabel(
    "Cases newly detected by M5\n"
    "among cases missed by M4 (%)",
    labelpad=8,
)
#ax.set_title("All of Us · 1-year landmark", fontsize=11, pad=10)

ax.grid(axis="y", color="0.90", linewidth=0.6)
ax.set_axisbelow(True)
ax.legend(loc="best", frameon=False, fontsize=9)

# ------------------------------------------------------------
# 5. Save using your notebook's existing functions
# ------------------------------------------------------------

save_gcs_csv(
    summary_s4,
    f"{RESULT_DIR}/"
    f"supp_S4_rescue_by_domain_coverage_1y_{FINAL_DATE}.csv",
)

save_figure_all(
    fig_s4,
    "Supplementary_Figure_S4_M4_M5_rescue_by_domain_coverage_1y",
    dpi=600,
)

plt.show()

print(
    f"M4 false negatives: {n_total:,}\n"
    f"Newly detected by M5: {n_rescued:,}\n"
    f"Overall recovery: {overall_rate:.2%}\n"
    f"Realized coverage groups: {len(summary_s4)}"
)

display(summary_s4)
```
